## **defaultdict**
**> A defaultdict is like a regular dictionary, except that when you try to look up a key it doesn’t contain, it first adds a value for it using a zero-argument function you provided when you created it**

```python
document = "The old grandfather clock in the hallway didn’t just tell time it seemed to breathe it Every rhythmic thud of the pendulum felt like a heartbeat anchoring the silent house against the rush of the outside world Sunlight filtered through the dusty windowpane illuminating a million dancing specks of gold that swirled whenever someone dared to walk past It was the kind of afternoon where the air felt thick with unspoken stories and even the tea cooling on the side table seemed to be waiting for a secret to be shared"

document
```

```python
words_count = {}
for word in document.split(sep=" "):
    if word in words_count:
        words_count[word] += 1
    else:
        words_count[word] = 1

words_count
```

```python
words_count = {}
for word in document.split():
    try:
        words_count[word] += 1
    except KeyError:
        words_count[word] = 1

words_count
```

```python
words_count = {}
for word in document.split():
    previosCount = words_count.get(word, 0)
    words_count[word] = previosCount + 1

words_count
```

```python
from collections import defaultdict

words_count = defaultdict(int) # creates a dictionary that returns 0 for keys that don't exist

for word in document.split():
    words_count[word] += 1

words_count
```

```python
dd_list = defaultdict(list) # creates a dictionary that returns an empty list for keys that don't exist
dd_list[2].append(8)

print(dd_list)
print(dd_list[3]) # returns an empty list since the key 3 doesn't exist in the dictionary
```

```python
dd_pair = defaultdict(lambda: [0, 0]) # creates a dictionary that returns a list [0, 0] for keys that don't exist
dd_pair[2][1] = 1

print(dd_pair)
print(dd_pair[3]) # returns [0, 0] since the key 3 doesn't exist in the dictionary
```

## **Object-Oriented Programming**

```python
class CountingClicker:
    """A clicker that counts the number of times it has been clicked."""

    # Constructor
    def __init__(self, count = 0): # init called "dunder" method contain double underscore
        # self is a reference to the instance of the class that is being created
        self.count = count
    
    # __repr__ method is a special method that is called when you print an instance of the class
    def __repr__(self):
        return f"CountingClicker(count={self.count})"
    
    def click(self, nbr_times = 1):
        """Increment the count by n times (default 1)"""
        self.count += nbr_times
    
    def read(self):
        """Return the current count"""
        return self.count
    
    def reset(self):
        """Reset the count to 0"""
        self.count = 0
```

```python
clicker1 = CountingClicker()
clicker2 = CountingClicker(100)
clicker3 = CountingClicker(count=100)

print("clicker1 count : ", clicker1.count)
print("clicker2 count : ", clicker2.count)
print("clicker3 count : ", clicker3.count)
```

**- Note :**

**> Class methods whose names start with an underscore are—by convention—considered “private,” and users of the class are not supposed to directly call them. However, Python will not stop users from calling them.**

```python
clicker = CountingClicker()
print(clicker)
assert clicker.read() == 0, "Initial count should be 0"
clicker.click()
assert clicker.read() == 1, "Count should be 1 after one click"
clicker.reset()
assert clicker.read() == 0, "Count should be 0 after reset"
```

```python
# subClass inherits all the methods and attributes of the parent class (CountingClicker in this case)
class NonResetClicker(CountingClicker):

    def reset(self):
        """Override the reset method to do nothing"""
        pass
```

```python
clicker = NonResetClicker()
print(clicker)
assert clicker.read() == 0, "Initial count should be 0"
clicker.click()
assert clicker.read() == 1, "Count should be 1 after one click"
clicker.reset()
assert clicker.read() == 1, "Count should still be 1 after reset since reset method is overridden to do nothing"
```

## **Iterables and Generators**

**> Think of a Python Generator as a "lazy" iterator. While a regular function computes its entire result and returns it all at once, a generator produces items one at a time, only when you ask for them.**

**> This makes them incredibly memory-efficient, especially when dealing with massive datasets.**

**2. Why Use Generators?**

**🚀 Memory Efficiency : If you need to process a list of 10 million integers, a standard list will store all 10 million in your RAM. A generator only stores the logic to create the next number, using almost zero memory regardless of the "list" size.**

**🐢 Lazy Evaluation : Generators don't do work until you force them to. If you only end up needing the first 5 items of a billion-item sequence, the generator only calculates those 5.**

```python
def generate_range(n):
    i = 0
    while i < n:
        # yield: Pauses the function, saves its state (all local variables), and sends a value back. 
        # When you call the generator again, it picks up exactly where it left off.
        yield i # yield is used to create a generator, it allows the function to return a value and pause 
                # its execution, and resume from the same point when the next value is requested
        i += 1
```

```python
for i in generate_range(5):
    print(i)
```

### **generator comprehension**

```python
evens_below_20 = (x for x in generate_range(20) if x % 2 == 0)

evens_below_20
```

```python
def count_up_to(n):
    count = 1
    while count <= n:
        yield count  # The function pauses here
        count += 1

# Using the generator
counter = count_up_to(3)
print(next(counter)) # Output: 1
print(next(counter)) # Output: 2
print(next(counter)) # Output: 3
```

## **zip and Argument Unpacking**
**> The zip function transforms multiple iterables into a single iterable of tuples**

```python
list1 = ["a", "b", "c"]
list2 = [1, 2, 3]

# zip() takes two or more iterables and returns an iterator of tuples, where the i-th tuple contains 
# the i-th element from each of the argument sequences or iterables. The iterator stops when the shortest 
# input iterable is exhausted.
pairs = [pair for pair in zip(list1, list2)]

pairs
```

```python
# * operator is used to unpack the list of pairs into separate arguments for the zip function, 
# effectively unzipping the pairs back into two separate lists
letters, number = zip(*pairs) # unzip the pairs into two separate lists

letters, number
```

```python
# Callable is a type from the typing module that represents a function with specified types for its
from typing import Callable

# Callable[[str, int], str] indicates that repeater is a function that takes a string and an integer 
# as arguments and returns a string
def twice(repeater: Callable[[str, int], str], s: str) -> str:
    return repeater(s, 2)

# This is an example of a function that matches the Callable type specified in the twice function.
def comma_repeater(s: str, n: int) -> str:
    n_copies = [s for _ in range(n)]
    return ', '.join(n_copies)

assert twice(comma_repeater, "type hints") == "type hints, type hints"
```
