## **Pure Functions**

A **pure function** is a concept in programming—specifically in "functional programming"—that describes a function that is completely predictable and self-contained.

For a function to be considered "pure," it must satisfy two strict rules:

- **1. Deterministic (Same Input = Same Output)**
    - A pure function must always return the exact same output if you give it the same input. It cannot rely on any data that might change outside of the function, such as global variables, system time, or random numbers.

* **Pure:** A function that adds two numbers together. `5 + 5` will always be `10`.
* **Impure:** A function that adds a number to the "current time." It will return a different result depending on when you run it.

---

- **2. No Side Effects**
    - A pure function must not change anything in the outside world. It should be "blind" and "deaf" to the rest of the application. It cannot:
        * Modify a global variable.
        * Print text to the console (this is a change to the outside world).
        * Write to a file or database.
        * Modify an object or list that was passed to it as an argument.

---

- **Why are Pure Functions useful?**
    * **Predictability:** Because they don't depend on external state, they are much easier to reason about. You don't have to worry about what else is happening in your program.
    * **Testability:** You can test them easily. You know exactly what the result will be, so you don't need to "mock" databases or external services.
    * **Concurrency:** Because they don't modify shared data, you can run pure functions in parallel without worrying about "race conditions" (where two parts of your code try to change the same variable at the same time).

---

- **Comparison: Pure vs. Impure**

| Feature | Pure Function | Impure Function |
| :--- | :--- | :--- |
| **Output** | Depends *only* on inputs | Can depend on external state |
| **Side Effects** | None (does not change anything) | Can modify variables/files/databases |
| **Reliability** | Highly predictable | Less predictable |
| **Testing** | Easy | Difficult |

---

- **An Analogy: The Coffee Machine**
    * **Pure Function:** You press the "Black Coffee" button. Every single time, you get a cup of black coffee. The machine doesn't care who you are, what time it is, or if the coffee shop down the street is closed.
    * **Impure Function:** You press the "Coffee" button, but the machine is connected to a sensor that checks if it's raining. If it's raining, it adds extra sugar. Or, the machine might suddenly run out of beans and decide to print an error message on the wall. The output changes based on things other than just pressing the button.

In modern Python programming, while you can't be 100% pure (because your program has to do *something* useful like saving a file), writing **more** of your code as pure functions makes your program significantly easier to maintain and debug.

> Functional programming is a paradigm that treats computation as the evaluation of **mathematical functions** and avoids changing-state and mutable data. While Python is primarily an object-oriented language, it has powerful tools to support a functional style.

- **The Core Pillars of Functional Style**

To write in a functional style in Python, focus on these three core principles:

- **1. Immutability**
    - Instead of modifying an existing list or dictionary, you create a new one. This prevents "spooky action at a distance" where your data changes unexpectedly elsewhere in your code.
* **Bad (Mutable):** `my_list.append(4)`
* **Good (Immutable):** `new_list = my_list + [4]`

---

- **2. Declarative Code**
    - You tell the computer **what** to do, rather than **how** to do it. Instead of writing `for` loops to iterate through items, you use high-level functions that describe the transformation.

---

- **3. Higher-Order Functions**
    - These are functions that take other functions as arguments or return them as results. Python makes this easy because functions are "first-class objects"—you can pass them around like variables.

---

- **Key Tools in Python's Functional Toolkit**

- **A. Map**
    - Applies a specific function to every item in an iterable.
```python
numbers = [1, 2, 3, 4]
# Square every number
squared = list(map(lambda x: x**2, numbers)) 
# Output: [1, 4, 9, 16]
```

---

- **B. Filter**
    - Creates a new collection containing only the items that pass a specific condition (a "predicate").
```python
numbers = [1, 2, 3, 4, 5, 6]
# Keep only even numbers
evens = list(filter(lambda x: x % 2 == 0, numbers))
# Output: [2, 4, 6]
```

---

- **C. Lambda Functions**
    - Small, anonymous, one-line functions. They are the "workhorses" of functional programming in Python.

---

- **D. List Comprehensions**
    - Often considered the "Pythonic" way to do functional programming. They act as a more readable version of `map` and `filter`.
```python
# The "Functional" way to do map and filter in one line
squared_evens = [x**2 for x in numbers if x % 2 == 0]
```

---

- **Comparison: Procedural vs. Functional**

| Feature | Procedural/Imperative | Functional |
| :--- | :--- | :--- |
| **Focus** | How to perform the task (steps) | What the output should be (transformation) |
| **State** | Shared, mutable state | No mutable state (immutability) |
| **Control Flow** | Loops (`for`, `while`) | Recursion and higher-order functions |
| **Functions** | Often rely on side effects | Pure functions only |

---

- **Should you use this for everything?**
    - Functional programming is excellent for data processing, concurrent programming, and reducing bugs in complex logic. However, Python is not a "pure" functional language like Haskell or Lisp. 

* **Use functional style for:** Data cleaning, processing lists, and mathematical transformations.
* **Use object-oriented style for:** Managing complex system state, UI components, and large-scale application architecture.

> Most excellent Python code is a **hybrid**—using functional style for data logic and object-oriented style for structural organization.

In the context of software development and system design, **prototype** and **patch** represent two distinct stages of a project's lifecycle: the "birth" of an idea and the "maintenance" of a finished product.

---

- **1. Prototype: The "Proof of Concept"**
    - A prototype is an early, experimental model of a system or product. Its primary goal is **exploration**. You build it to test if an idea works, to gather feedback, or to demonstrate feasibility to stakeholders.

* **Characteristics:**
    * **Rough/Incomplete:** It often lacks error handling, security, or full-scale performance.
    * **Disposable:** It is often written to be thrown away or heavily refactored once the core concepts are proven.
    * **Speed-Focused:** The priority is to build it as fast as possible to validate assumptions.

---

- **2. Patch: The "Correction"**
    - A patch is a small piece of software (or a code snippet) designed to **update, fix, or improve** an existing, deployed program. Its primary goal is **remediation**.

* **Characteristics:**
    * **Specific:** It targets a known bug, security vulnerability, or minor feature deficiency.
    * **Stable:** It must be carefully tested because it is being applied to a system that is already "live" and being used by people.
    * **Additive:** It essentially "patches" a hole in the existing fabric of the application.

---

- **The Strategic Difference**

| Feature | Prototype | Patch |
| :--- | :--- | :--- |
| **Stage** | Beginning of lifecycle | Middle/End of lifecycle |
| **Purpose** | To learn if an idea is viable | To fix or improve existing work |
| **Risk** | High (intended to be messy) | High (can break stable systems) |
| **Expectation** | To be iterated or discarded | To be integrated permanently |

---

- **How They Work Together**
    - In a healthy development cycle, these two concepts often dance together:
        1.  You build a **prototype** to test a feature.
        2.  Once validated, you build the **production-grade** version.
        3.  Later, as bugs are found or requirements change, you release **patches** to keep that production version running safely and effectively.

---

- **A "Real World" Analogy**
    * **The Prototype** is like a **sketch on a napkin** from an architect showing the general layout of a house. It proves the layout works, but you wouldn't try to live in the napkin.
    * **The Patch** is like **replacing a leaky faucet** in your finished, furnished house. You aren't rebuilding the house; you are fixing one specific, broken part so the whole system continues to function.

## **Design-First Development**

**Design-First Development** is an architectural philosophy that prioritizes defining the structure, contracts, and interfaces of a system **before** writing the actual implementation code.

Instead of writing code and then figuring out how different components talk to each other, you treat the "blueprint" as the source of truth.

---

- **1. The Core Philosophy**
    - In a standard "Code-First" approach, you might start writing a Python function or a database table immediately. In "Design-First," you start with a **contract**. This contract acts as the communication agreement between the client (who uses the service) and the server (who provides it).

* **The Artifacts:** You create files like OpenAPI (Swagger) specs for APIs, Protocol Buffers for gRPC, or ER diagrams for database schemas.
* **The Tools:** You use these specs to **generate** the boilerplate code, documentation, and even test cases automatically.

---

- **2. Why is this becoming popular?**
    - Design-First is the gold standard for **Microservices** and distributed systems. 

* **Parallel Development:** Once the contract (the API spec) is finished, the backend team can start implementing the server, and the frontend team can start building the UI against a "mocked" version of that contract—they don't have to wait for the backend to be "done."
* **Documentation is First-Class:** The documentation isn't an afterthought; it *is* the design. The spec file effectively acts as the manual for your system.
* **Consistency:** By using automated generators, you ensure that every service in your organization follows the same naming conventions, error-handling structures, and data types.

---

- **3. The "Design-First" Workflow**
    1.  **Contract Definition:** Write the API spec (e.g., `openapi.yaml`).
    2.  **Mocking:** Use a tool (like Prism or Mirage) to generate a "fake" server that follows the contract.
    3.  **Frontend/Backend Development:** Both teams work simultaneously based on the spec.
    4.  **Implementation:** The backend team writes the code that actually fulfills the contract.
    5.  **Validation:** Automated tests check the code against the original spec to ensure no one accidentally "broke the contract."

---

- **4. Pros and Cons**

| Feature | Design-First | Code-First |
| :--- | :--- | :--- |
| **Speed** | Slower at the start (upfront design). | Very fast to start coding. |
| **Collaboration** | Excellent; teams are aligned early. | Can lead to silos or "breakages." |
| **Documentation** | Always up to date (it's the source). | Often drifts from implementation. |
| **Best For** | Complex, multi-team projects. | Prototypes, small scripts, solo projects. |

---

- **The "Architectural" Analogy**
    * **Code-First** is like building a house by just starting to lay bricks and deciding where the windows go as you build. It’s fast if you’re building a shed, but it’s a nightmare if you’re building a skyscraper.
    * **Design-First** is like hiring an architect to produce detailed blueprints. You spend weeks on paper before a single brick is laid. It takes longer to start, but when you build, you know exactly how the plumbing, electricity, and structure will fit together, and you can have the plumbing and electrical teams working at the same time without hitting each other.

````python
def time_to_int(time):
    minute = time.hour * 60 + time.minute
    second = minute * 60 + time.second

    return second
````

````python
start = make_time(1, 1, 1)

print_time_lunch(start)
````

````python
time_to_int(start)
````

````python
def int_to_time(second):
    minute, second = divmod(second, 60)
    hour, minute = divmod(minute, 60)

    return make_time(hour, minute, second)
````

````python
time = int_to_time(3661)

print_time_lunch(time)
````

````python
def add_time(time, hour, minute, second):
    duration = make_time(hour, minute, second)
    second = time_to_int(time) + time_to_int(duration)

    return int_to_time(second)
````

## **Debugging**

- **The Toolkit for Object Inspection**

Here is how you can use these tools to diagnose your objects:

- **1. `type(object)`**
    - This is your most basic diagnostic. If you have an object and aren't sure what "kind" of thing it is, `type` will tell you the class it belongs to.
        * **Example:** If you call `type(start)`, it might return `__main__.Time`, telling you the object is an instance of the `Time` class defined in your current script.

---

- **2. `isinstance(object, class)`**
    - Instead of asking "What are you?", this asks "Are you one of these?". It is a more robust way to check if an object is compatible with a certain class, as it also accounts for class inheritance.
        * **Example:** `isinstance(end, Time)` returns `True` if `end` is indeed a `Time` object.

---

- **3. `hasattr(object, 'attribute')`**
    - One of the most common causes of bugs is trying to access an attribute that doesn't exist (e.g., trying to read `start.second` when you only defined `hour` and `minute`). `hasattr` lets you "look before you leap."
        * **Example:** `hasattr(start, 'hour')` returns `True` if the attribute exists, allowing you to avoid a `AttributeError`.

---

- **4. `vars(object)`**
    - This is the most powerful tool for seeing exactly what is inside your object. It returns a dictionary containing the object's attributes and their current values.
        * **Example:** `vars(start)` might give you `{'hour': 9, 'minute': 40, 'second': 0}`. This is perfect for seeing if your object's internal data matches what you *think* it should be.

---

- **5. `structshape(object)`**
    - When you have complex data structures (like a tuple containing several custom objects), `structshape` helps you visualize the "shape" of the data. It gives you a summary of the structure so you can verify that your nested data is organized correctly.
        * **Example:** `structshape(t)` might return `'tuple of 2 Time'`, confirming that you have a tuple containing two `Time` objects.

---

- **Summary Checklist for Debugging**

| Function | Best Used For |
| :--- | :--- |
| **`type()`** | Finding the class name of an object. |
| **`isinstance()`** | Verifying if an object is the expected class. |
| **`hasattr()`** | Checking if an attribute exists before using it. |
| **`vars()`** | Peering inside the object to see all current values. |
| **`structshape()`** | Visualizing the hierarchy of complex or nested structures. |



> **Pro-Tip:** When you get an `AttributeError` or find that your calculations are wrong, the first step should always be calling `vars(obj)`. It usually reveals exactly which attribute is missing or holds the wrong value in seconds.

class Time:

obj = Time()

print(type(obj))

#### **What is the difference between a variable and an attribute?**

The difference between a variable and an attribute is essentially about **scope and ownership**. While both store data, they "live" in different places and belong to different contexts within your code.

---

- **1. Variables (The "Local" or "Global" Storage)**
    - A variable is a name assigned to a value that exists within a specific scope, such as inside a function, a module, or the global program. It is independent and does not necessarily "belong" to any specific object.

* **Where they live:** They exist in the namespace of the function or script currently being executed.
* **Ownership:** They are "free-floating." They are not bound to an object.
* **Example:**
```python
    x = 10  # This is a variable
    def my_function():
        y = 5  # This is a local variable inside the function
```

---

- **2. Attributes (The "Object" Property)**
    - An attribute is a variable that is **attached to an object**. It is a piece of data that describes a property of that specific object. If you have a `Car` object, the "color" or "model" are attributes of that car.

* **Where they live:** They live inside the object (specifically, in the object's `__dict__`).
* **Ownership:** They are bound to an instance of a class. You access them using "dot notation" (`object.attribute`).
* **Example:**
```python
    class Car:
        def __init__(self):
            self.color = 'red'  # 'color' is an attribute of the Car instance

    my_car = Car()
    print(my_car.color)  # Accessing the attribute using dot notation
```

---

- **The Key Differences**

| Feature | Variable | Attribute |
| :--- | :--- | :--- |
| **Ownership** | Independent / Scope-bound | Owned by an object |
| **Access** | Accessed by name (e.g., `x`) | Accessed via dot notation (e.g., `obj.x`) |
| **Lifetime** | Exists as long as the scope is active | Exists as long as the object exists |
| **Purpose** | Temporary storage / local calculations | Describing the state of an object |

---

- **The "Label" Analogy**
    - Imagine you are in a workshop:
        * **A Variable** is like a **sticky note** you hold in your hand. You can write a number on it, change it, or throw it away when you're done with a specific calculation. It doesn't belong to any particular tool in the room.
        * **An Attribute** is like a **label permanently attached** to a specific drill. The label says "Battery Level: 80%." That data is part of the drill itself. Even if you put the drill away, the label (attribute) stays with it because it describes the drill's state.

> **Think of it this way:** If you can access it by just saying the name (`x`), it’s a variable. If you have to say "that object's [something]" (`object.x`), it’s an attribute.

#### **What are the pros and cons of pure functions compared to modifiers?**

In programming, particularly when discussing object-oriented or stateful systems, there is a fundamental tension between **Pure Functions** (which perform calculations without changing state) and **Modifiers** (methods or patterns that actively change the internal state of an object).

To understand the comparison, we must define the "modifier" pattern: a modifier (or "mutator") is a function that changes the attributes of an object, typically returning `None` because its purpose is the *side effect* of changing data.

---

- **1. Pure Functions**
    - A pure function takes an input, performs a calculation, and returns a *new* result without modifying the original data.

- **Pros:**
    * **High Predictability:** Because there are no side effects, you never have to worry that calling a function will accidentally change the value of an object elsewhere in your program.
    * **Easier Debugging:** If the output is wrong, the bug is inside the function. You don't have to look at the global state or the object's history.
    * **Concurrency-Friendly:** Multiple threads can call pure functions simultaneously without risking "race conditions," since no shared state is being changed.

- **Cons:**
    * **Memory Overhead:** Since pure functions return *new* data rather than modifying the existing object, they can consume more memory, as you are frequently creating copies of data.
    * **Verbosity:** You often have to manage the passing of state manually (e.g., passing an object into a function, getting a new one back, and replacing the old one).

---

- **2. Modifiers (Mutators)**
    - A modifier function directly alters the attributes (the "state") of an existing object.

- **Pros:**
    * **Memory Efficiency:** Modifying an object in place is very efficient. You don't need to create a new object in memory; you simply update the existing values.
    * **Natural for Real-World Modeling:** Objects often represent things that *do* change over time. It is often more intuitive to say `car.speed = 100` than to create a whole new `Car` object with a different speed value.
    * **Simpler Syntax:** You don't have to worry about reassigning variables; you just call the method and the object updates itself.

- **Cons:**
    * **Hidden Side Effects:** If you pass an object into a function and that function modifies it, the original object changes—often leading to bugs where other parts of your program receive unexpected data.
    * **Difficult to Track:** Tracking *when* and *where* a complex object’s state changed can become a nightmare in large, multi-threaded applications.

---

- **Comparison Summary**

| Feature | Pure Functions | Modifiers (Mutators) |
| :--- | :--- | :--- |
| **State** | Immutable (no change) | Mutable (changes state) |
| **Efficiency** | Lower (creates new objects) | Higher (modifies in place) |
| **Safety** | High (no side effects) | Lower (risk of hidden bugs) |
| **Best Used For** | Mathematical logic, data processing | Large data structures, performance-critical tasks |

---

- **Which one should you use?**
    - The modern best practice is a **Hybrid Approach**:
        1.  **Favor Pure Functions** for your business logic, data transformation, and calculations. This keeps the "brain" of your application clean and testable.
        2.  **Use Modifiers** (or setters) specifically when you need to represent entities that inherently have a lifecycle or when you are dealing with very large datasets where creating copies would significantly slow down your application.

> **The Golden Rule:** Always try to isolate your modifiers. Keep the parts of your code that change things (the modifiers) separate from the parts that calculate things (the pure functions).

#### **What format specifiers can be used in a Python f-string?**

Python f-strings (introduced in Python 3.6) use the same powerful formatting mini-language as the `.format()` method. You trigger this by placing a colon `:` after the expression inside the curly braces.

Here is a breakdown of the most common format specifiers.

- **1. Numeric Formatting**
    - These are essential for controlling precision, padding, and base representation.

* **Decimal Precision (`:.nf`):** Controls how many digits appear after the decimal point.
```python
    value = 3.14159
    print(f"{value:.2f}")  # Output: '3.14'
```
* **Separators (`:,`):** Adds thousands separators.
    ```python
    large_num = 1000000
    print(f"{large_num:,}")  # Output: '1,000,000'
    ```
* **Percentages (`:.n%`):** Multiplies by 100 and adds a `%` sign.
```python
    rate = 0.756
    print(f"{rate:.1%}")  # Output: '75.6%'
```

---

- **2. Padding and Alignment**
    - Useful for creating clean tables or formatted console output. The syntax is `:[fill][align][width]`.

* **Alignment (`<`, `>`, `^`):** Left, Right, and Center alignment.
* **Width:** The total number of characters the output should occupy.
* **Fill:** An optional character to fill the empty space (default is space).

```python
name = "Gemini"
print(f"|{name:<10}|")  # Left align, width 10: '|Gemini    |'
print(f"|{name:>10}|")  # Right align, width 10: '|    Gemini|'
print(f"|{name:^10}|")  # Center align, width 10: '|  Gemini  |'
print(f"|{name:*^10}|") # Center align with '*' fill: '|**Gemini**|'
```

---

- **3. Date and Time Formatting**
    - If you are working with `datetime` objects, f-strings support the standard `strftime` directives.

* **Format codes:** `%Y` (Year), `%m` (Month), `%d` (Day), `%H` (Hour), `%M` (Minute).
```python
    from datetime import datetime
    now = datetime.now()
    print(f"{now:%Y-%m-%d %H:%M}") # Output: '2026-04-19 15:18'
```

---

- **4. Other Useful Specifiers**

* **Binary, Octal, Hex (`:b`, `:o`, `:x`):** Converts integers into different number bases.
```python
    num = 255
    print(f"{num:x}")  # Output: 'ff' (hexadecimal)
    print(f"{num:b}")  # Output: '11111111' (binary)
```
* **Scientific Notation (`:e`):**
```python
    big = 1230000000
    print(f"{big:.2e}")  # Output: '1.23e+09'
```

---

- **Quick Reference Summary**

| Specifier | Description | Example |
| :--- | :--- | :--- |
| `:.2f` | Float with 2 decimal places | `3.14` |
| `:,` | Thousands separator | `1,000` |
| `:<10` | Left-aligned, width 10 | `'text      '` |
| `:>10` | Right-aligned, width 10 | `'      text'` |
| `:^10` | Centered, width 10 | `'   text   '` |
| `:.1%` | Percentage, 1 decimal place | `'50.0%'` |
| `:x` | Hexadecimal | `'ff'` |

### **Exercise 1**

- Write a function called subtract_time that takes two Time objects and returns the interval between them in seconds—assuming that they are two times during the same day.

````python
class Time:
    """classe that represente the current tiem of day"""
````

````python
def make_time(hour, minute, second):
    obj_time = Time()

    res , obj_time.second = divmod(second, 60)
    res, obj_time.minute = divmod(minute + res, 60)
    _, obj_time.hour = divmod(hour + res, 60)

    return obj_time

def print_time(obj_time):
    print(f'{obj_time.hour:0d}:{obj_time.minute:0d}:{obj_time.second:0d}')


def time_to_int(obj_time):
    second = obj_time.second + (obj_time.minute * 60) + (obj_time.hour * 3600)

    return second

def subtract_time(obj_time1, obj_time2):
    time1 = time_to_int(obj_time1)
    time2 = time_to_int(obj_time2)

    return abs(time1 - time2)
````

````python
time1 = make_time(9, 42, 1)
time2 = make_time(11, 32, 5)

print_time(time1)
print_time(time2)
````

````python
subtract_time(time1, time2)
````
