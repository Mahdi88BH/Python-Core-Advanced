In Python, a **List** is a versatile container used to store multiple items in a single variable. Imagine it as a physical "to-do list" or a physical folder where you can keep a collection of different objects together in a specific order.

---

## **Core Characteristics of Lists**

- **1. Ordered Sequence**
    - Lists maintain a strict order. Every item added to a list has a specific position (known as an index). The first item is always at the beginning, the second is next, and so on. This order does not change unless you specifically tell Python to move or remove an item.

- **2. Changeable (Mutable)**
    - Unlike strings, lists are **mutable**. This means you can add, remove, or modify items within the list after it has already been created. You can swap the third item for something else without having to recreate the entire list from scratch.

- **3. Allows Duplicates**
    - A list can contain the same value multiple times. If you want to store the number "5" ten times in a row, a list will allow it and keep track of each individual instance based on its position.

- **4. Heterogeneous (Mixed Types)**
    - Lists are extremely flexible; they do not care what kind of data you put inside them. A single list can simultaneously hold:
        * Whole numbers (Integers)
        * Text (Strings)
        * Decimal numbers (Floats)
        * Even other lists!

- **5. Dynamic Size**
    - You do not need to define how big a list will be when you create it. Lists can grow or shrink automatically as you add or delete items.

---

- **Summary Comparison**

| Characteristic | Description |
| :--- | :--- |
| **Ordered** | Items have a fixed place and rank. |
| **Mutable** | The contents can be edited freely. |
| **Non-Unique** | Duplicate values are perfectly fine. |
| **Flexible** | Can store any mix of data types together. |

## **List Methods**

![Python-list-methods](./Images/Python-list-methods.png)

- **1. Adding Items**
    - These methods make your list grow.

* **`append()`**: Adds a single item to the very **end** of the list.
    ```python
    fruits = ['apple', 'banana']
    fruits.append('cherry') # ['apple', 'banana', 'cherry']
    ```
* **`extend()`**: Adds **multiple** items (from another list or iterable) to the end.
    ```python
    nums = [1, 2]
    nums.extend([3, 4]) # [1, 2, 3, 4]
    ```
* **`insert()`**: Adds an item at a **specific position** (index).
    ```python
    colors = ['red', 'blue']
    colors.insert(1, 'green') # ['red', 'green', 'blue']
    ```


---

- **2. Removing Items**
    - These methods make your list shrink.

* **`remove()`**: Searches for the **first occurrence** of a specific value and deletes it.
    ```python
    pets = ['cat', 'dog', 'cat']
    pets.remove('cat') # ['dog', 'cat']
    ```
* **`pop()`**: Removes and **returns** the item at a specific index. If you don't provide an index, it removes the last item.
    ```python
    letters = ['a', 'b', 'c']
    last = letters.pop() # last is 'c', list is now ['a', 'b']
    ```
* **`clear()`**: Removes **every single item** from the list, leaving it empty.
    ```python
    data = [1, 2, 3]
    data.clear() # []
    ```

---

- **3. Finding and Counting**
    - These methods give you information about what's inside without changing the list.

* **`index()`**: Returns the position (index) of the first item with the specified value.
    ```python
    chars = ['x', 'y', 'z']
    print(chars.index('y')) # 1
    ```
* **`count()`**: Returns the number of times a specific value appears in the list.
    ```python
    grades = ['A', 'B', 'A', 'C']
    print(grades.count('A')) # 2
    ```

---

- **4. Organizing the List**
    - These methods rearrange the order of the items.

* **`sort()`**: Sorts the list in ascending order (alphabetical for strings, numerical for numbers). This changes the original list.
    ```python
    prices = [50, 10, 30]
    prices.sort() # [10, 30, 50]
    ```
* **`reverse()`**: Flips the order of the list (the first item becomes the last).
    ```python
    tasks = ['start', 'middle', 'end']
    tasks.reverse() # ['end', 'middle', 'start']
    ```

---

- **5. Utility**
* **`copy()`**: Returns a **shallow copy** of the list. This is useful because if you just say `list_b = list_a`, changing one will change the other. Using `.copy()` prevents this.
    ```python
    original = [1, 2, 3]
    new = original.copy()
    ```


---

- **Summary Cheat Sheet**

| Action | Method | Modifies List? |
| :--- | :--- | :--- |
| **Add to end** | `append()` | Yes |
| **Add many to end** | `extend()` | Yes |
| **Add at index** | `insert()` | Yes |
| **Remove by value** | `remove()` | Yes |
| **Remove by index** | `pop()` | Yes |
| **Find index** | `index()` | No |
| **Count occurrences** | `count()` | No |
| **Alphabetize/Order** | `sort()` | Yes |

## **Objects and Values**

````python
a = 'mahdi'
b = 'mahdi'
````

- To check whether two variables refer to the same object, you can use the is operator

````python
a is b
````

- But when you create two lists, you get two objects

````python
a = [1, 2, 3]
b = [1, 2, 3]

print(f"are a and b refere to the same object? => {a is b}")
````

> In this case we would say that the two lists are **equivalent** because they have the same elements, but are not **identical** because they are not the same object. If two objects are **identical**, they are also **equivalent**, but if they are equivalent, they are not necessarily identical.

This is one of the most fascinating "under the hood" behaviors in Python. It comes down to two concepts: **String Interning** and the difference between **Immutable** vs. **Mutable** objects.

---

- **1. Strings: The "Interning" Optimization**
    - In Python, strings are **immutable** (they cannot be changed). Because of this, Python tries to save memory through a process called **Interning**.

When you create the string `'mahdi'`, Python looks in its memory "bank" to see if that exact string already exists. Since it does, Python simply points the second variable (`b`) to the same memory address as the first (`a`).

**Why does Python do this?**
    * **Memory Efficiency:** Why store the same text twice?
    * **Speed:** Comparing if two variables point to the same object ($a \text{ is } b$) is much faster than checking every single character in a long string.

---

- **2. Lists: The "Identity" Requirement**
    - Lists are **mutable** (you can change them). This completely changes the rules. 
    - Even if `l1` and `l2` contain the exact same numbers right now, Python **must** create two separate objects in memory.

- **Why can't lists be interned?**
    - Imagine if Python pointed `l1` and `l2` to the same object to save memory. Look what would happen:
        1.  You decide to add a number to `l1`: `l1.append(4)`.
        2.  If they shared the same object, `l2` would suddenly have a `4` in it too!

This would cause massive, unpredictable bugs. Therefore, every time you use square brackets `[]`, Python says: *"I'm going to build a brand-new container in a brand-new spot in memory."*

---

- **3. "Equal" vs. "The Same"**
    - In Python, we use two different operators to check these relationships:
        * **`==` (Equality):** Checks if the **data** inside is the same.
            * `l1 == l2` is `True` (They both have `[1, 2, 3]`).
        * **`is` (Identity):** Checks if they are the **exact same object** in memory.
            * `l1 is l2` is `False`.
            * `a is b` is `True`.

---

- **The Summary Table**

| Data Type | Property | Shared in Memory? | Reason |
| :--- | :--- | :--- | :--- |
| **String** | Immutable | **Yes** (usually) | Interning saves space safely. |
| **List** | Mutable | **No** | Changing one shouldn't break the other. |

> **Note:** Python only "interns" strings that look like identifiers (letters, numbers, underscores) and are relatively short. If you create a very long string with special characters, Python might skip the interning and create two objects!

Does that help you see why Python treats "containers" differently than "values"?

## **List Arguments**

- This is a fundamental concept in programming often called passing by reference (or more specifically in Python, passing by object reference).

- Unlike a simple integer or string, which is typically copied when passed to a function, a list is a complex object. When you pass it to a function, you aren't sending a copy of the list; you’re sending the "address" or "link" to the original list in memory.

**How it Works**
- Think of the list as a physical notebook. Instead of photocopying every page of the notebook and giving the copy to the function, you are giving the function the key to the drawer where the original notebook is kept. Anything the function writes in that notebook is permanently there when you open the drawer later.

````python
def pop_elements(lst):
    return lst.pop(0)
````

````python
letters = ['a', 'b', 'c']

# the parameter lst and the variable letters are aliases 

# for the same object
for _ in range(len(letters)):
    print(pop_elements(letters))
````

````python
letters
````

**Why does this happen?**
- **Efficiency:** Lists can be massive (thousands of items). Copying the entire list every time you pass it to a function would slow down your computer and eat up memory.
- **Mutable vs. Immutable:** Lists are mutable, meaning they are designed to be changed "in place." Simple types like numbers ($5$, $10$) or strings ("Hello") are immutable; you can't change them, you can only create new ones.
**The "Gotcha": Reassignment vs. Modification**
- It is important to distinguish between modifying the list and reassigning the variable.
    - **Modification:** `lst.pop()`, `lst.append()`, or `lst[0] = "New"` changes the original.
    - **Reassignment:** If you say lst = [1, 2, 3] inside the function, you have broken the link to the original "notebook" and created a brand new one. The caller will not see this change.
> **Pro-Tip:** If you want a function to work on a list without changing the original, you should pass a copy of it instead: `pop_first(my_fruits[:])`. The [:] syntax creates a shallow copy.

This concept is a cornerstone of how memory management works in languages like Python, Java, or JavaScript. When you pass an object (like a list, dictionary, or custom class instance) to a function, you aren't passing a "photocopy" of that object—you are passing the **memory address** of the actual object.

---

#### **Understanding Aliasing**
**Aliasing** occurs when more than one variable name points to the same spot in your computer's memory. 

Imagine you have a house. If you give a friend a piece of paper with your address on it, they now have a "reference" to your house. 
* Your friend (the function) doesn't have a *different* house; they have an **alias** for yours. 
* If they paint your front door red, the door is red when you get home because you both share the same physical building.



---

#### **Why Changes "Persist"**
Because the function is working on the exact same data structure stored in memory, any **mutations** (changes) it makes are permanent. The "link" between the caller's variable and the function's parameter ensures they are both looking at the same data.

#### **The Code Perspective**
```python
def add_item(target_list):
    # 'target_list' is an alias for 'my_data'
    target_list.append("New Item") 

my_data = ["Old Item"]
add_item(my_data)

print(my_data) 
# Output: ["Old Item", "New Item"]
```

In this example, `target_list` and `my_data` are aliases. When the function finishes, the local variable `target_list` is destroyed, but the **change** it made to the list in memory remains.

---

#### **The Persistence Table**
Here is how different actions affect the "Persistence" of changes:

| Action | Example | Does it Persist? | Why? |
| :--- | :--- | :--- | :--- |
| **Modification** | `obj.append(x)` | **Yes** | You changed the contents of the memory address. |
| **Mutation** | `obj[0] = 5` | **Yes** | You updated a value inside the existing object. |
| **Reassignment** | `obj = [1, 2]` | **No** | You pointed the alias to a *new* address; the original is untouched. |

---

#### **When to Be Careful**
Aliasing can lead to "Side Effects," which are bugs where data changes unexpectedly because a function modified an object you didn't want it to touch. 

> **Tip:** If you want a function to manipulate data without changing the original, you should pass a **copy** of the object: 
> `my_function(my_list.copy())`

### **Exercise1**

- Two words are anagrams if you can rearrange the letters from one to spell the other.For example, tops is an anagram of stop. One way to check whether two words are anagrams is to sort the letters in both words. If the lists of sorted letters are the same, the words are anagrams.
- Write a function called is_anagram that takes two strings and returns True if they are anagrams. Using your function and the word list, find all the anagrams of takes.

````python
def is_anagram(word1, word2):
    return sorted(word1) == sorted(word2)
````

````python
anagram_words = []
words_list = open('./words.txt').read().splitlines()
target_word = 'listen'

for element in words_list:
    if element != target_word and len(element) == len(target_word):
        if is_anagram(element, target_word):
            anagram_words.append(element)

print("anagram words are => ", anagram_words)
````
