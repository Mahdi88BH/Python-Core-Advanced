In Python, a **Tuple** is a collection used to store multiple items in a single variable, very similar to a list, but with one major, defining difference: **permanence**.

Think of a tuple as a **sealed envelope** or a "read-only" list. Once you put the information inside and close the envelope, you cannot change, add, or remove anything without creating a whole new envelope.

---

## **Core Characteristics of Tuples**

- **1. Ordered Sequence**
    - Like lists, tuples maintain a strict order. Every item has a specific position (index) starting from zero. If you put "North" in the first slot and "South" in the second, they will stay in those exact spots forever.

- **2. Immutable (Unchangeable)**
    - This is the "Superpower" of the tuple. Once a tuple is created, it is **constant**. You cannot append new items, remove existing ones, or change the value of an item at a specific index. This makes tuples safer for data that should never be altered by your program (like the coordinates of a city or the RGB values of a color).

- **3. Allows Duplicates**
    - Since items are tracked by their position, you can have the same value appear multiple times. A tuple of `(1, 1, 1)` is perfectly valid.

- **4. Heterogeneous (Mixed Types)**
    - Tuples are flexible containers. A single tuple can hold a mix of strings, integers, floats, and even other tuples or lists.

- **5. Faster and More Memory Efficient**
    - Because Python knows that a tuple will never change, it can optimize how it stores it in your computer's memory. This makes tuples slightly faster to process than lists and takes up less space.



---

- **Summary Comparison: Tuple vs. List**

| Characteristic | Tuple | List |
| :--- | :--- | :--- |
| **Mutability** | **Immutable** (Unchangeable) | **Mutable** (Changeable) |
| **Syntax** | Uses Parentheses `()` | Uses Square Brackets `[]` |
| **Performance** | Faster / Memory Efficient | Slower / Heavier |
| **Best For** | Fixed data (Constants, Coordinates) | Changing data (Shopping lists, Scores) |

---

- **Why use a Tuple instead of a List?**
If you can do almost everything with a list, why bother with tuples? 
1.  **Safety:** It prevents accidental changes to data that should stay constant.
2.  **Dictionary Keys:** Because tuples are unchangeable, they can be used as **keys** in a dictionary, whereas lists cannot.
3.  **Clarity:** Using a tuple tells other programmers: *"This data is a fixed group; don't try to edit it."*

## **Tuple methods**

Because tuples are **immutable** (unchangeable), they do not have methods that add, remove, or change items. You won't find `append()`, `remove()`, or `sort()` here.

In fact, Python tuples only have **two** built-in methods! Both of these methods are strictly for *finding information* about what is already inside the tuple.

---

- **1. `count()`**
    - This method searches the tuple and returns the number of times a specific value appears.

- **How it works:**
    - You give it a value, and it scans the tuple from left to right, keeping a tally of how many exact matches it finds. If the value isn't in the tuple at all, it simply returns `0`.

**Example:**
```python
# A tuple of grades
grades = ('A', 'B', 'A', 'C', 'A')

# Count how many 'A's are in the tuple
total_As = grades.count('A')
print(total_As)  # Output: 3

# Try to count a grade that isn't there
total_Fs = grades.count('F')
print(total_Fs)  # Output: 0
```

---

- **2. `index()`**
    - This method searches the tuple for a specific value and returns the **position (index)** where it was found.

- **How it works:**
    - You give it a value, and it scans from left to right. As soon as it finds the *very first* match, it stops and tells you that index number (remember, Python starts counting at 0). 

*Important Note:* If the value you are looking for does not exist in the tuple, Python will crash with a `ValueError`.

**Example:**
```python
# A tuple of racing placements
placements = ('gold', 'silver', 'bronze', 'silver')

# Find the index of 'bronze'
bronze_position = placements.index('bronze')
print(bronze_position)  # Output: 2

# Find the index of 'silver'
# Notice it stops at the first one it sees!
first_silver = placements.index('silver')
print(first_silver)  # Output: 1
```

---

- **Summary**

| Method | What it does | What happens if the value isn't there? |
| :--- | :--- | :--- |
| **`count(x)`** | Returns how many times `x` appears. | Returns `0`. |
| **`index(x)`** | Returns the position of the *first* `x`. | Crashes with a `ValueError`. |

Because tuples are designed to be lightweight, read-only data containers, these two methods provide exactly what you need to inspect the data without risking accidental modifications.

## **Argument Packing**

- Functions can take a variable number of arguments. A parameter name that begins with the **\* operator packs** arguments into a tuple.

````python
def mean(*args):
    return sum(args) / len(args)
````

````python
mean(1, 2, 3, 4, 5)
````

- If you have a sequence of values and you want to pass them to a function as multiple arguments, you can use the * operator to **unpack** the tuple.

````python
t = (5, 3)
divmod(*t)
````

````python
t = [5, 3]
divmod(*t)
````

````python
def min_max(t):
    return min(t), max(t)

def mean(*args):
    return sum(args) / len(args)

def trimmed_mean(*args):
    min, max = min_max(args)
    trimmed = list(args)
    trimmed.remove(min)
    trimmed.remove(max)

    return mean(*trimmed)
````

````python
min_max([1, 2, 3, 10])
````

````python
trimmed_mean(1, 2, 3, 10)
````

## **Inverting a Dictionary**

````python
def letter_counter(word):
    counter = {}
    for letter in word:
        if letter in counter:
            counter[letter] += 1
        else:
            counter[letter] = 1
    return counter
````

````python
letter_counter('parrot')
````

````python
def invert_dict(word):
    counter = letter_counter(word)
    new_dict = {}

    for key, value in counter.items():
        if value not in new_dict:
            new_dict[value] = [key]
        else:
            new_dict[value].append(key)
    
    return new_dict
        
````

````python
invert_dict('parrot')
````

### **Ask a Virtual Assistant**

#### **Are Python tuples always hashable?**

The short answer is: **No, not always.**

While tuples are **immutable** (they cannot be changed), being immutable is not the only requirement for being **hashable**. To be hashable, an object must also contain only other hashable objects.

---

- **1. The Rule of Thumb**
    - In Python, an object is hashable if it has a hash value that never changes during its lifetime.
        * **Hashable:** Integers, floats, strings, and booleans.
        * **Unhashable:** Lists, dictionaries, and sets (because they are mutable).

Because a tuple is a container, its "hashability" depends entirely on its contents.

---

- **2. The "Pure" Tuple (Hashable)**
    - If a tuple contains only immutable objects (like strings, integers, or other "pure" tuples), it is **hashable**. This means it can be used as a key in a dictionary or added to a set.

```python
# All elements are immutable
point = (10, 20) 
locations = {point: "Home"} # Works perfectly!
```

---

- **3. The "Hybrid" Tuple (Unhashable)**
    - If a tuple contains even one **mutable** object (like a list), the tuple itself becomes **unhashable**. 

Even though you can't replace the list inside the tuple with a different list, the *contents* of that list can change. If the list changes, the hash of the tuple would have to change too—and that violates the rules of hashing.

```python
# Contains a mutable list
bad_tuple = (1, 2, [3, 4]) 

# This will raise a TypeError: unhashable type: 'list'
my_set = {bad_tuple} 
```
---

- **4. Why does this matter?**
    - The most common place this "gotcha" appears is when using dictionaries.
        * **Dictionaries** use hashes to find values instantly. 
        * If you try to use a tuple containing a list as a dictionary key, Python stops you because it can't guarantee that the "key" will stay the same forever. If the list inside the tuple changed, Python would "lose" the data associated with that key.

---

- **Summary Checklist**

| Content of Tuple | Is the Tuple Hashable? | Can it be a Dict Key? |
| :--- | :--- | :--- |
| `(1, 2, "hello")` | **Yes** | **Yes** |
| `(1, 2, (3, 4))` | **Yes** | **Yes** |
| `(1, 2, [3, 4])` | **No** | **No** |
| `(1, 2, {"a": 1})` | **No** | **No** |

Think of it this way: A tuple is like a chain. The chain is only as "strong" (hashable) as its weakest (mutable) link.

### **Exercise 2**

- Write a function called most_frequent_letters that takes a string and prints the letters in decreasing order of frequency.

````python
def selecting(t):
    return t[1]

def most_frequent_letters(word):
    letters = []
    counter = []

    for char in word:
        if char not in letters:
            letters.append(char)
            counter.append(word.count(char))
    
    return [element[0] for element in sorted(zip(letters, counter), key=selecting, reverse=True)]
````

````python
most_frequent_letters('banana')
````

````python
def most_frequent_letters_opt_version(word):
    letter_count = {}
    for char in word:
        letter_count[char] = letter_count.get(char, 0) + 1
    
    return sorted(letter_count, key=letter_count.get, reverse=True)
````

````python
most_frequent_letters_opt_version('banana')
````

- “Metathesis” is the transposition of letters in a word. Two words form a “metathesis pair” if you can transform one into the other by swapping two letters, like converse and conserve. Write a program that finds all of the metathesis pairs in the word list.

````python
def in_anagram(word1, word2):
    return word1[::-1] == word2[::-1]

def distence_between(*word):
    
    counter = 0
    for m, n in zip(*word):
        
        if m != n:
            counter += 1
    return counter


def is_matathesis(list_word):
    
    anagram_word = {}
    matathesis_pairs = []

    for word in list_word:
        if ''.join(sorted(word)) in anagram_word:
            anagram_word[''.join(sorted(word))].append(word)
        else:
            anagram_word[''.join(sorted(word))] = [word]

    for value in anagram_word.values():
        if len(value) == 2:
            if distence_between(*value) == 2:
                matathesis_pairs.append(value)
    
    return matathesis_pairs
````
