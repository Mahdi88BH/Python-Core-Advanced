## **Debugging**

- **1. The `doctest` Approach**
This method is best for simple, quick checks. You write the expected behavior directly inside the function's docstring.

```python
from doctest import run_docstring_examples

def add(a, b):
    """
    Add two numbers.
    >>> add(2, 2)
    4
    >>> add(10, 5)
    15
    """
    return a + b

# This runs the tests found inside the docstring
def run_doctests(func):
    run_docstring_examples(func, globals(), name=func.__name__)

# Execution
run_doctests(add)  # No output means all tests passed
```

---

- **2. The `unittest` Approach**
This approach is more structured and professional. It uses a class-based system to organize tests and provides detailed feedback if something goes wrong.

```python
import unittest

# The function we want to test
def add(a, b):
    return a + b

# Define a class that inherits from unittest.TestCase
class TestExample(unittest.TestCase):
    
    # Test method MUST start with 'test_'
    def test_add(self):
        result = add(2, 2)
        # self.assertEqual checks if result == 4
        self.assertEqual(result, 4)
        
    def test_add_fail_example(self):
        # This will trigger a failure report because 2+2 is not 100
        self.assertEqual(add(2, 2), 100)

# Helper function to run the tests
def run_unittest():
    # argv=[''] prevents it from trying to read command line arguments
    # exit=False prevents the test runner from closing the notebook/program
    unittest.main(argv=[''], verbosity=2, exit=False)

# Execution
run_unittest()
```
---

- **Key Takeaways for your Debugging Strategy:**
    * **When to use `doctest`**: Use this when you are writing small utility functions or when you want your documentation to double as a simple test.
    * **When to use `unittest`**: Use this for larger projects. Because it handles **Test Discovery** (automatically finding all methods starting with `test_`), you can keep adding new tests without having to manually call each one individually.
    * **The "Fail" Report**: Note that when `unittest` fails, it provides a `Traceback`. This is critical for debugging because it tells you:
        * **Which** test method failed.
        * **Where** in the code it failed (the exact line).
        * **What** went wrong (e.g., `AssertionError: 4 != 100`).

Python’s `set` class provides a powerful set of tools for handling unordered collections of unique elements. You can interact with sets using both **built-in methods** (which often modify the set in-place or return a new set) and **operators** (which provide a concise, readable syntax for set logic).

- **1. Set Operators**
    - Operators are often preferred for their readability when performing mathematical set operations between two set objects (e.g., `s1 | s2`).

| Operation | Operator | Description |
| :--- | :--- | :--- |
| **Union** | `\|` | Returns a new set with elements from both sets. |
| **Intersection** | `&` | Returns a new set with elements common to both sets. |
| **Difference** | `-` | Returns a new set with elements in the first set but not the second. |
| **Symmetric Difference** | `^` | Returns a new set with elements in either set, but not both. |
| **Subset** | `<=` | Returns `True` if the set is a subset of the other. |
| **Superset** | `>=` | Returns `True` if the set is a superset of the other. |

    - Methods provide more granular control and often support any iterable (like lists or tuples) as an argument, whereas operators usually require both sides to be set objects.

- **Adding and Removing Elements**
    * **`add(elem)`**: Adds a single element to the set.
    * **`remove(elem)`**: Removes the specified element; raises a `KeyError` if not found.
    * **`discard(elem)`**: Removes the specified element; does **not** raise an error if not found.
    * **`pop()`**: Removes and returns an arbitrary element from the set.
    * **`clear()`**: Removes all elements from the set.

- **Set Logic Methods**
    - These are the functional equivalents of the operators listed above:
        * **`union(*others)`**: Returns a new set with elements from the set and all others.
        * **`intersection(*others)`**: Returns a new set with elements common to the set and all others.
        * **`difference(*others)`**: Returns a new set with elements in the set that are not in the others.
        * **`symmetric_difference(other)`**: Returns a new set with elements in either the set or the other, but not both.

- **Relationship Testing**
    * **`isdisjoint(other)`**: Returns `True` if the two sets have no elements in common.
    * **`issubset(other)`**: Returns `True` if every element of the set is in the other.
    * **`issuperset(other)`**: Returns `True` if every element of the other is in the set.

#### **What are the methods and operators of Python’s counter class?**

Python's `Counter` class, found in the `collections` module, is a specialized subclass of `dict` designed for counting hashable objects. While it inherits all standard dictionary methods (like `.keys()`, `.values()`, and `.items()`), it adds powerful tools specifically for tallying frequencies.

---

- **1. Counter-Specific Methods**
    - These methods are unique to `Counter` and provide efficient ways to manipulate and analyze the frequency counts.

| Method | Description |
| :--- | :--- |
| **`elements()`** | Returns an iterator over elements, repeating each as many times as its count. |
| **`most_common(n)`** | Returns a list of the `n` most common elements and their counts, sorted from most to least frequent. |
| **`total()`** | Computes the sum of all counts in the Counter. |
| **`update(iterable)`** | Adds counts from another iterable or mapping (unlike `dict.update`, which replaces them). |
| **`subtract(iterable)`** | Subtracts counts from another iterable or mapping. |

---

- **2. Arithmetic and Set Operators**
    - `Counter` supports mathematical and set-like operations that treat the counts as part of the logic. This makes it extremely useful for comparing datasets or tallying groups.

| Operation | Operator | Description |
| :--- | :--- | :--- |
| **Addition** | `+` | Adds two counters, keeping only positive results. |
| **Subtraction** | `-` | Subtracts the second counter from the first, keeping only positive results. |
| **Intersection** | `&` | Returns a new Counter containing the **minimum** of corresponding counts. |
| **Union** | `\|` | Returns a new Counter containing the **maximum** of corresponding counts. |

---

- **Key Behaviors to Note:**
    * **Missing Keys:** Unlike a standard dictionary that raises a `KeyError`, accessing a missing key in a `Counter` returns **0**.
    * **Positive Counts Only:** Operations like `+`, `-`, `&`, and `|` automatically filter out elements with zero or negative counts, keeping the Counter clean.
    * **Unary Operations:** You can use `+counter` to remove all zero or negative counts from a `Counter` object.

- **Simple Example**
```python
from collections import Counter

# Initialization
c = Counter(['a', 'a', 'b', 'c', 'c', 'c'])

# Most common
print(c.most_common(2)) 
# Output: [('c', 3), ('a', 2)]

# Math operations
c2 = Counter(a=1, b=2)
result = c + c2
print(result)
# Output: Counter({'c': 3, 'a': 3, 'b': 3})
```

#### **What is the difference between a Python list comprehension and a generator expression?**

The core difference between a list comprehension and a generator expression is **how they handle data in memory**. While they look nearly identical syntactically, they serve different performance needs.

---

- **1. List Comprehension**
    - A list comprehension creates the **entire list in memory** as soon as it is executed.

* **Syntax:** Uses square brackets `[]`.
* **Behavior:** It eagerly evaluates every expression, computes every item, and stores them in a list.
* **Best for:** Small-to-medium datasets where you need to access specific elements by index (e.g., `my_list[0]`) or need to iterate over the data multiple times.

```python
# Creates the whole list of 1 million items in memory immediately
squares = [x**2 for x in range(1000000)]
```

---

- **2. Generator Expression**
    - A generator expression creates an **iterator object**. It does not compute the values immediately; it calculates them **on the fly** (lazily) as you iterate over them.

* **Syntax:** Uses parentheses `()`.
* **Behavior:** It yields one item at a time only when requested. Once an item is yielded, it is discarded from memory.
* **Best for:** Large datasets, data streams, or scenarios where you only need to iterate through the data once. It is much more memory-efficient.

```python
# Creates a generator object; memory usage is constant regardless of size
squares_gen = (x**2 for x in range(1000000))

# Values are computed only as you ask for them
print(next(squares_gen)) # 0
print(next(squares_gen)) # 1
```

---

- **Summary Comparison**

| Feature | List Comprehension | Generator Expression |
| :--- | :--- | :--- |
| **Syntax** | `[x for x in data]` | `(x for x in data)` |
| **Memory Usage** | High (stores all elements) | Low (stores one element at a time) |
| **Evaluation** | Eager (all at once) | Lazy (one by one) |
| **Access** | Supports indexing (e.g., `[0]`) | No indexing; only iteration |
| **Performance** | Faster if data is small/reused | Faster if data is large/streamed |

---

- **When to choose which?**
    * **Use a List Comprehension** if you need the full data collection, if you need to slice or index it, or if the collection is small enough that memory isn't a concern.
    * **Use a Generator Expression** if you are processing massive amounts of data (like reading a multi-gigabyte file line by line) or if you only need to iterate over the sequence once. Using a generator in these cases prevents your program from consuming all your system's RAM.

#### **When should I use Python’s namedtuple rather than define a new class?**

A `namedtuple` is a factory function from the `collections` module that creates a lightweight, immutable object with named fields. Deciding between a `namedtuple` and a full class comes down to your needs for **mutability, behavior, and complexity**.

---

- **When to use `namedtuple`**
    - `namedtuple` is an excellent choice when you want the benefits of a class—specifically readable, attribute-based access—without the memory overhead or boilerplate code of a full class definition.

* **You need simple data containers:** If your object is primarily a "bucket of data" without complex behavior or methods.
* **You need immutability:** `namedtuple` objects are immutable (like tuples). Once created, you cannot change the values, which prevents accidental side effects in your code.
* **You want memory efficiency:** `namedtuple` is much more memory-efficient than a standard class because it does not have a `__dict__` for each instance.
* **You want to leverage tuple behavior:** Because it is still a tuple, it supports unpacking (e.g., `x, y = point`) and indexing.

---

- **When to define a new Class**
    - A standard class is necessary when you need features that `namedtuple` simply cannot provide.

* **You need mutability:** If you need to modify object attributes after creation, a standard class is required.
* **You need custom methods:** While you *can* technically add methods to a `namedtuple` via inheritance, it is awkward. If your object needs to perform complex actions or maintain internal state, a standard class is the standard approach.
* **You need data validation or default values:** If you need to enforce complex rules upon initialization, or if you need to set default values for attributes, standard classes are more flexible.
* **You are building a complex hierarchy:** If you need inheritance, polymorphism, or complex class-based architecture, `namedtuple` will quickly become a limitation.

---

- **Quick Comparison Table**

| Feature | `namedtuple` | Standard Class |
| :--- | :--- | :--- |
| **Mutability** | Immutable | Mutable (usually) |
| **Memory** | Very Low | Higher |
| **Syntax** | Concise | Verbose |
| **Primary Use** | Data structures/Records | Objects with logic/behavior |
| **Access** | Attributes & Indexing | Attributes |

---

- **Modern Alternative: `dataclasses`**
    - If you find yourself needing the best of both worlds—the concise syntax of a `namedtuple` but the flexibility of a class—use Python's `dataclasses` (introduced in Python 3.7+).

```python
from dataclasses import dataclass

@dataclass
class Point:
    x: int
    y: int

# Pros:
# 1. Mutable by default (but can be frozen).
# 2. Supports default values.
# 3. Much cleaner than a traditional class.
# 4. Supports methods easily.
```

**Recommendation:** Use `namedtuple` for fixed, read-only data records (like database rows). Use `dataclasses` for objects that require state, default values, or potential mutability.

#### **What are some uses of packing and unpacking keyword arguments?**

Packing and unpacking keyword arguments (often called `**kwargs`) in Python is a powerful pattern that allows you to write more flexible, extensible, and clean code.

---

- **1. Packing: Capturing Arbitrary Arguments**
    - **Packing** uses `**` in a function definition to collect all passed keyword arguments into a single dictionary.

* **Use Case: Middleware or Decorators.** When you are writing a wrapper function (like a decorator) that needs to pass arguments through to another function without knowing what those arguments are ahead of time.
* **Use Case: Configuring Objects.** When you want to accept any number of configuration settings without defining each one individually in the function signature.

```python
def log_function_call(func, **kwargs):
    print(f"Calling function: {func.__name__}")
    print(f"Arguments: {kwargs}")
    return func(**kwargs)

# Here, 'name' and 'age' are packed into the 'kwargs' dictionary
log_function_call(print, name="Alice", age=30)
```

---

- **2. Unpacking: Distributing Arguments**
    - **Unpacking** uses `**` in a function call to "spread" the contents of a dictionary into a function as individual keyword arguments.

* **Use Case: Passing Configuration Dictionaries.** If you have a dictionary of settings or parameters, you can pass them into a function instantly.
* **Use Case: Merging and Updating.** Easily updating object states or passing settings down a chain of function calls.

```python
def create_user(name, age, email):
    print(f"Creating user {name} ({age}) - {email}")

config = {"name": "Bob", "age": 25, "email": "bob@example.com"}

# Unpacking the dictionary into the function arguments
create_user(**config)
```

---

- **Summary of Benefits**

| Feature | Primary Benefit |
| :--- | :--- |
| **Flexibility** | Functions can accept optional parameters without being hard-coded. |
| **Conciseness** | Avoids writing large functions with many parameters (`f(a, b, c, d, e, ...)`). |
| **Extensibility** | Allows child classes or decorators to pass arguments to parent classes/wrapped functions without needing to update signatures. |

---

- **Best Practices & Pitfalls**
    * **Explicit is better than implicit:** While `**kwargs` is powerful, overusing it makes your code harder to read because the function signature doesn't tell the user what arguments are actually expected. If a function always requires specific arguments, list them explicitly.
    * **Overriding:** When unpacking a dictionary into a function that also has standard arguments, the dictionary keys must match the parameter names, or the function will raise a `TypeError`.
    * **Combining:** You can use `*args` (for positional arguments) and `**kwargs` together, provided they are in the order `def func(f_args, *args, **kwargs):`.

#### **How does unittest do test discovery?**

`unittest` performs test discovery by traversing your project's directory structure and using introspection to identify test code. This mechanism allows you to run your entire suite of tests simply by calling the runner, without having to manually list every file or test class.

---

Here is how the process works step-by-step:

- **1. Identifying the "Test" Modules**
When you invoke discovery—often via `python -m unittest discover` or `unittest.main()`—the framework acts like a file system crawler:

* **Starting Point:** It starts at a designated directory (defaulting to the current working directory).
* **Pattern Matching:** It recursively searches for files that match a specific pattern. By default, this is `test*.py`.
* **Importability:** It attempts to import these files as Python modules. If a file cannot be imported (due to syntax errors or missing dependencies), `unittest` will typically log an error and skip it.

---

- **2. Finding the Test Classes**
    - Once `unittest` has imported a valid module, it inspects that module to find classes. It looks for **subclasses** of `unittest.TestCase`.

* **Introspection:** It uses the `dir()` function and `issubclass()` check on every object found in the module to identify which ones inherit from `TestCase`.

---

- **3. Discovering the Test Methods**
    - After identifying the relevant classes, `unittest` inspects the methods within those classes:
        * **Naming Convention:** It filters for methods that begin with the string **`test`**.
        * **Method Collection:** It creates a `TestSuite` object. For every class found, it iterates through its methods and adds any method starting with `test` to that suite as a unique test case.

---

- **Summary of the Discovery Logic**

| Step | Action | Logic |
| :--- | :--- | :--- |
| **Search** | Crawl files | Find all `test*.py` files recursively. |
| **Import** | Load modules | Turn files into Python objects for inspection. |
| **Verify** | Find `TestCase` | Identify classes inheriting from `unittest.TestCase`. |
| **Filter** | Match methods | Collect methods starting with `test`. |
| **Run** | Execute | Add all discovered tests to a `TestSuite` and run them. |

- **Why this is useful**
    - Because `unittest` uses this automated approach, you don't need to update a central list of tests every time you add a new one. As long as you follow the naming conventions (`test_...`), your new tests are automatically included in the next run. 

> **Important Note:** If you place your tests in a subdirectory, ensure that directory contains an `__init__.py` file (in older Python versions) or is structured as a package, so that `unittest` can correctly import the files during its discovery crawl.

#### **Along with assertequal, what are the most commonly used methods in unit test.TestCase?**

Beyond `assertEqual`, the `unittest.TestCase` class provides a robust library of assertion methods to handle various types of data validation. These methods are designed to provide descriptive error messages when a test fails, making it much easier to pinpoint exactly why your code didn't behave as expected.

---

- **1. Equality and Identity Assertions**
    - These are the building blocks of most tests.

* **`assertNotEqual(a, b)`**: Fails if `a == b`. Used to ensure a function does *not* return a specific unwanted value.
* **`assertIs(a, b)`**: Fails if `a` is not the same object as `b` (using `is`). Useful when testing singleton patterns or checking if a variable references the exact same object in memory.
* **`assertIsNot(a, b)`**: Fails if `a` is the same object as `b`.

- **2. Boolean and Truthiness Assertions**
    - These simplify checks against constants or conditional logic.

* **`assertTrue(x)`**: Fails if `bool(x)` is `False`.
* **`assertFalse(x)`**: Fails if `bool(x)` is `True`.
* **`assertIsNone(x)`**: Fails if `x` is not `None`.
* **`assertIsNotNone(x)`**: Fails if `x` is `None`.

---

- **3. Membership and Collection Assertions**
    - These are perfect for verifying contents of lists, dictionaries, and sets.

* **`assertIn(member, container)`**: Fails if `member` is not found in `container` (e.g., list, tuple, string).
* **`assertNotIn(member, container)`**: Fails if `member` is found in `container`.
* **`assertListEqual(list1, list2)`** (and `assertDictEqual`, `assertSetEqual`): Similar to `assertEqual` but provides a much more detailed diff if the collections differ.

---

- **4. Exception Assertions**
    - One of the most powerful features of `unittest` is verifying that your code **correctly raises errors** when it receives bad input.

* **`assertRaises(exception, callable, *args, **kwargs)`**: Fails if the `callable` does not raise the expected `exception`.

```python
# Example of testing an exception
def divide(a, b):
    return a / b

class TestMath(unittest.TestCase):
    def test_divide_by_zero(self):
        # Passes if divide(10, 0) raises ZeroDivisionError
        with self.assertRaises(ZeroDivisionError):
            divide(10, 0)
```

---

- **5. Numerical and String Comparison**
    - For more specific data types:
        * **`assertAlmostEqual(a, b, places=7)`**: Fails if the difference between `a` and `b` is greater than a specified precision. Crucial for floating-point math where `0.1 + 0.2 != 0.3` exactly.
        * **`assertRegex(text, regex)`**: Fails if the string `text` does not match the provided regular expression.

---

- **Summary Checklist for Assertions**

| Category | Method |
| :--- | :--- |
| **Equality** | `assertEqual`, `assertNotEqual` |
| **Identity** | `assertIs`, `assertIsNot` |
| **Boolean** | `assertTrue`, `assertFalse`, `assertIsNone` |
| **Membership** | `assertIn`, `assertNotIn` |
| **Exceptions** | `assertRaises` |
| **Precision** | `assertAlmostEqual` |



> Using the correct assertion method is better than just using `assertEqual` for everything, because when a test fails, `unittest` will provide a tailored error message (e.g., "3 is not in [1, 2]" instead of just "False != True").

### **Exercice 2**

- Write a function that takes a string of letters and a word, and checks whether the letters can spell the word, taking into account how many times each letter appears.

````python
from collections import Counter
````

````python
def scrabble_spell(sub_word, word):
    count_word = Counter(word)
    count_sub_word = Counter(sub_word)
    return all(count_sub_word[key] == count_word[key] for key in count_sub_word)
````

````python
scrabble_spell('BELT', 'TABLE')
````

````python
scrabble_spell('BEET', 'TABLE')
````

The following is a function that recursively computes the binomial coefficient:
```python
    def binomial_coeff(n, k):
    """Compute the binomial coefficient "n choose k".
    n: number of trials
    k: number of successes
    returns: int
    """
    if k == 0:
        return 1
    if n == 0:
        return 0
    return binomial_coeff(n-1, k) + binomial_coeff(n-1, k-1)
```
Rewrite the body of the function using nested conditional expressions.

````python
memo = {}  # This dictionary stores our results: key = (n, k), value = result

def binomial_coeff_memo(n, k):
    # 1. Look-up: Before doing any math, check if we've solved this before.
    # This turns an exponential time complexity problem into a linear one.
    if (n, k) in memo:
        return memo[(n, k)]
    
    # 2. Logic & Assignment: We use 'setdefault' to compute the value
    # only if the key doesn't exist, then store it in 'memo' automatically.
    return memo.setdefault(
        (n, k), 
        # Base Case 1: n choose 0 is always 1
        1 if k == 0 else 
        # Base Case 2: If we have trials but n is 0, we can't have successes
        (0 if n == 0 else 
        # Recursive Step: Pascal's Identity: C(n,k) = C(n-1, k) + C(n-1, k-1)
        binomial_coeff_memo(n-1, k) + binomial_coeff_memo(n-1, k-1))
    )
````
