## **Defining Methods**

````python
class Time():
    """class for present the time in the current day"""

    def __init__(self, hour=0, minute=0, second=0):
        self.hour = hour
        self.minute = minute
        self.second = second

    def print_time(self):
        print(f'{self.hour:02d}:{self.minute:02d}:{self.second:02d}')
    
    def __str__(self):
        s = f'{self.hour:02d}:{self.minute:02d}:{self.second:02d}'
        return s

    def time_to_int(self):
        return self.hour * 3600 + self.minute * 60 + self.second
    
    def int_to_time(second): # static method
        minute, second = divmod(second, 60)
        hour, minute = divmod(minute, 60)

        return make_time(hour, minute, second)
    
    def add_time(self, hour, minute, second):
        duration = make_time(hour, minute, second)
        seconds = self.time_to_int() + duration.time_to_int()

        return Time.int_to_time(seconds)
    
    def is_after(self, other):
        return self.time_to_int() > other.time_to_int()
    
    def __add__(self, other):
        second = self.time_to_int() + other.time_to_int()

        return Time.int_to_time(second)
````

````python
def make_time(hour, minute, second):
    time = Time()

    time.hour = hour
    time.minute = minute
    time.second = second

    return time
````

## **Debugging**

This section of the chapter introduces the concept of **Invariants** and the defensive programming technique of **Assertion**.

- **1. What is an Invariant?**
    - An **invariant** is a rule that **must always be true** for an object to be considered "valid." If this rule is broken, your program enters an invalid state, which will inevitably lead to bugs later on.

- For a `Time` object, the invariants are:
    * **Time must be non-negative:** Hours, minutes, and seconds must be $\ge 0$.
    * **Ranges:** Minutes and seconds must be less than 60 (since they represent standard clock time).
    * **Data Types:** Hours and minutes must be integers (though we might allow seconds to have fractions).

---

- **2. The Defensive Method: `is_valid()`**
- Instead of just hoping your objects are created correctly, you write an `is_valid()` method. This acts as a **sanity check**. It evaluates the internal data of your `Time` object and returns `True` if all invariants are met, or `False` if something is wrong.

* **Why use it?** It centralizes the rules. If you ever decide that a "Time" should actually allow up to 90 minutes, you only have to change the logic in one place.

---

- **3. Enforcing Rules with `assert`**
- The `assert` statement is the "policeman" of your code. It is used to verify that an assumption is correct at a specific point in time.

* **How it works:** `assert [condition], [error message]`
    * If the condition is **True**, the program continues silently.
    * If the condition is **False**, the program **crashes immediately** with an `AssertionError`.

- **Why would you *want* your program to crash?**
    - It is much better for a program to fail immediately with a clear error message (like `"self is not a valid Time"`) than to continue running with "bad data," which could produce incorrect results or corrupt files later on. By crashing, you find the bug at the source rather than chasing it through your logs hours later.

---

- **Summary Table**

| Concept | Purpose | Analogy |
| :--- | :--- | :--- |
| **Invariant** | Define "correct" data. | The rules of the game. |
| **`is_valid()`** | Check if rules are met. | The referee checking the players. |
| **`assert`** | Force a stop if rules are broken. | The referee blowing the whistle to halt play. |

- **Why use `assert` instead of `if`?**
- You could use an `if` statement to check validity, but `assert` is specifically designed for **debugging**. It clearly separates your "business logic" (what the program is actually supposed to do) from your "error checking" (making sure the program is healthy). In many Python environments, you can even disable all `assert` statements in production to speed up your code once you are confident it is bug-free.

#### **What’s the difference between an instance method and a static method?**

In Python, the difference between an **instance method** and a **static method** comes down to **what the method has access to** and **how it is called**.

---

- **1. Instance Method (The Default)**
An instance method is a function that belongs to a specific instance of a class. When you define it, the first parameter is almost always `self`, which represents the specific object you are working with.

* **Access:** It has full access to the object's attributes (`self.attribute`) and other methods.
* **Purpose:** To perform actions that modify or use the state of that specific instance.
* **Calling:** You call it on an object: `my_object.my_method()`.

---

- **2. Static Method (`@staticmethod`)**
    - A static method is a function that belongs to a class but doesn't need to know anything about the instance or the class itself. It’s like a regular function that just happens to be organized inside a class for convenience.

* **Access:** It **cannot** access `self` or `cls`. It is completely isolated from the instance's state.
* **Purpose:** To perform utility tasks that are related to the class conceptually but don't need to change any data.
* **Calling:** You can call it on the class itself: `MyClass.my_static_method()`.

---

- **Comparison Table**

| Feature | Instance Method | Static Method |
| :--- | :--- | :--- |
| **Decorator** | None (default) | `@staticmethod` |
| **First Argument** | `self` (the object) | None |
| **Access to State** | Can modify instance attributes | Cannot access instance or class state |
| **Best Used For** | Logic involving object data | Utility functions (math, string formatting) |

---

- **Code Example**
```python
class Calculator:
    def __init__(self, brand):
        self.brand = brand

    # Instance method: needs 'self' to know which calculator it is
    def display_brand(self):
        print(f"This is a {self.brand} calculator.")

    # Static method: just a helper function; doesn't care about the brand
    @staticmethod
    def add(a, b):
        return a + b

# Usage:
calc = Calculator("Casio")
calc.display_brand()      # Calling instance method
print(Calculator.add(5, 3)) # Calling static method
```

---

- **Which one should you use?**
    * **Use an Instance Method** if your code needs to read or write data stored in `self`. If the logic changes based on the specific "thing" you are working on, it’s an instance method.
    * **Use a Static Method** if your code performs a calculation or transformation that doesn't depend on the object's internal state. If you find yourself thinking, "This function is related to my class, but it doesn't actually use any of the class's data," it should likely be a static method.

#### **Why are static methods called static?**

The term "static" comes from the world of computer memory and language design. While modern Python abstracts much of this away, the name persists because it describes **how the method behaves in relation to the memory of the program.**

---

- **1. The "Static" vs. "Dynamic" Distinction**
    - In older languages like C or C++, the term "static" refers to something that is **fixed at compile-time** rather than **allocated at run-time**.

* **Dynamic (Instance Methods):** When you create an object (an instance), you are dynamically allocating memory for that object and its data. An instance method is "dynamic" because it works specifically on that unique piece of memory.
* **Static (Static Methods):** A static method is **"static"** because its memory address and logic are fixed the moment the program starts. It does not need to know about the "dynamic" objects (the instances) created while the program is running. It exists independently of any instance.

---

- **2. The "Static" Nature of the Method**
    - Because static methods don't depend on `self` (the instance) or `cls` (the class), they are effectively **constant** or **immobile**. 

* **They don't move or change based on context:** An instance method's behavior changes depending on the data inside the object (the `self`). A static method always does the exact same thing, regardless of what object called it or what the object's state is.
* **They don't "bind":** In Python, instance methods are "bound" to an object—they are connected to a specific instance at runtime. Static methods are "unbound"; they are just functions that have been tucked away inside a class for organizational purposes.

---

- **3. Historical Context**
    - The term is borrowed from languages like **Java and C++**, where `static` is a keyword. In those languages, a `static` member is shared by all instances of a class and lives in a special area of memory that is allocated only once for the entire application, rather than once for every object created.

Even though Python is much more flexible, it kept the name "static" because it correctly signals to programmers coming from other languages: **"This method does not need an instance to work, and it will not change based on instance data."**

---

- **Summary Table**

| Perspective | Why it is called "Static" |
| :--- | :--- |
| **Memory** | Its definition is fixed in memory; it doesn't need to be created alongside an object. |
| **Behavior** | It is "stationary"—it doesn't change its internal state or rely on context. |
| **Scope** | It is global to the class rather than local to the instance. |


In essence, a static method is "static" because it is **detached from the lifecycle of the object.** It is a piece of code that stands still, unchanging and independent, while your objects come and go in the program's memory.

#### **what is a Decorators?**

In Python, a **decorator** is a design pattern that allows you to "wrap" another function to extend its behavior without permanently modifying its source code.

Think of a decorator as a **wrapper** or a **layer of paint**. You aren't changing the structure of the house (the original function), but you are changing how it looks or behaves from the outside.

---

- **1. How it works**
    - A decorator is a function that:
        1. Takes another function as an input.
        2. Defines a new function inside itself (a "wrapper") that adds some extra logic.
        3. Returns that new wrapper function.

- **2. Basic Syntax**
    - You apply a decorator using the `@` symbol right above the function definition.

```python
# The Decorator
def my_decorator(func):
    def wrapper():
        print("Something is happening before the function is called.")
        func()
        print("Something is happening after the function is called.")
    return wrapper

# Applying the Decorator
@my_decorator
def say_hello():
    print("Hello!")

# Calling the function
say_hello()
```

---

- **3. Why are they useful?**
    - Decorators are primarily used to separate **"cross-cutting concerns"** from your main business logic. If you have 20 different functions and you want to log every time any of them is called, you don't want to manually add a `print()` statement inside every one of them. You write *one* decorator and apply it to all 20.

- **Common real-world use cases:**
    * **Logging:** Recording when a function was called.
    * **Authentication:** Checking if a user is logged in before allowing them to run a function.
    * **Timing:** Measuring how long a function takes to execute.
    * **Caching (Memoization):** Storing the results of expensive function calls so they don't have to be recalculated.

---

- **4. Key Takeaways**

| Feature | Description |
| :--- | :--- |
| **Input** | A function. |
| **Output** | A new "wrapped" function. |
| **Persistence** | Original function code is never touched. |
| **Cleanliness** | Keeps your code "DRY" (Don't Repeat Yourself). |



- **The "Gift Wrapping" Analogy**
    - Imagine you have a book (your original function). You want to give it as a gift, so you wrap it in paper and add a bow (the decorator). 
        * The **book** is still just a book; it doesn't know it's a gift.
        * The **wrapper** adds the "gift-giving" behavior. 
        * When someone receives the gift, they interact with the *wrapper* first, then get to the book. You didn't have to rewrite the book to make it a gift; you just added an external layer.

#### **what is utility of self?**

In Python, the `self` parameter is the **bridge between the blueprint (the class) and the specific object (the instance).** When you write a class, you are writing code for an object that doesn't exist yet. `self` is a placeholder that Python automatically replaces with the *actual* object when the code runs.

---

- **1. The Core Utility: Accessing "My" Data**
    - The primary utility of `self` is to allow an instance method to distinguish between "general" variables and the data **stored inside that specific object.**

Without `self`, a method wouldn't know which object's data it is supposed to be working on.

```python
class Dog:
    def __init__(self, name):
        self.name = name  # 'self.name' is the attribute for THIS specific dog

    def bark(self):
        print(f"{self.name} says woof!") # 'self' tells Python: "use the name of this dog"

dog1 = Dog("Buddy")
dog2 = Dog("Rex")

dog1.bark() # self becomes dog1; prints "Buddy says woof!"
dog2.bark() # self becomes dog2; prints "Rex says woof!"
```

---

- **2. Why it is required in Python**
    - Unlike some other languages (like Java or C++) where the connection to the instance is implicit, Python requires you to be **explicit**. 

* **Explicit is better than implicit:** By requiring you to write `self` as the first argument, Python makes it crystal clear that the method is an instance method and that it will be operating on an object's state. 
* **Flexibility:** Because `self` is just an argument, you can technically name it anything (like `this` or `me`), though doing so is strongly discouraged as it violates Python community standards.

---

- **3. Utility in Method Chaining and Communication**
    - `self` also allows methods to talk to each other within the same object. If you have a method that needs to call another method belonging to the same instance, you use `self`.

```python
class Robot:
    def power_on(self):
        print("Powering on...")
    
    def greet(self):
        self.power_on()  # 'self' allows this method to call another method on the same object
        print("Hello!")
```

---

- **4. Summary of Utility**

| Utility | Description |
| :--- | :--- |
| **Identification** | Allows the code to reference the current instance's attributes. |
| **Isolation** | Keeps data scoped to a specific object (prevents data leakage). |
| **Communication** | Allows methods to call other methods on the same object. |
| **Explicit Logic** | Makes it clear to the programmer that the method is bound to an object. |



- **The "ID Card" Analogy**
    - Think of `self` as the object's **ID Card**. 
        * When a method is called, the Python "system" looks at which object is calling it.
        * It hands that object’s "ID Card" (`self`) to the method.
        * Now, the method has access to all the information on that specific ID Card (the attributes like `name`, `age`, etc.), so it knows exactly which object it is serving.

Without `self`, the method would be like a worker in a factory who doesn't have a clipboard; they would have no idea which product they are currently working on.

1. Write a definition for a Date class that represents a date—that is, a year, month, and day of the month.
2. Write an **`__init__`** method that takes year, month, and day as parameters and assigns the parameters to attributes. Create an object that represents June 22, 1933.
3. Write a **`__str__`** method that uses a format string to format the attributes and returns the result. If you test it with the Date you created, the result should be 1933-06-22.
4. Write a method called is_after that takes two Date objects and returns True if the first comes after the second. Create a second object that represents September 17, 1933, and check whether it comes after the first object.

````python
class Date():
    """Represent a date in the following format %Y-%m-%d"""

    def __init__(self, year, month, day):
        self.year= year
        self.month= month
        self.day = day
    
    def __str__(self):
        return f'{self.year}-{self.month:02d}-{self.day:02d}'
    
    def date_to_int(self):
        return self.year * 365 + self.month * 30 + self.day
    
    def is_after(self, other):
        return self.date_to_int() > other.date_to_int()
````

````python
date2 = Date(1933, 9, 17)

date1.is_after(date2)
````
