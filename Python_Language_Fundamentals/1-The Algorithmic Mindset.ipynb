## **Expression**

- The Standard Hierarchy
> When an expression contains multiple operators, they are evaluated in the following sequence:
1. **Parentheses / Brackets:** Anything inside $( )$ or $[ ]$ is solved first
2. **Exponents / Orders:** Powers and square roots (e.g., $2^3$, $\sqrt{9}$).
3. **Multiplication and Division:** These hold equal rank and are processed from left to right.
4. **Addition and Subtraction:** These also hold equal rank and are processed from left to right.

## **Formal and Natural Languages**

1. **Ambiguity**
> **Ambiguity** occurs when a single expression, statement, or command can be interpreted in more than one way.
- **In Language:** "I saw a man with a telescope." (Did I use the telescope, or did the man have it?)
- **In Programming:** If the order of operations isn't clear, an expression like $10 + 5 * 2$ is ambiguous to a human unless they know the language's specific precedence rules.
- **The Goal:** High-quality code and data schemas aim for zero ambiguity so the computer always performs exactly what the developer intended.
2. **Redundancy**
> **Redundancy** refers to the inclusion of information that is unnecessary or already exists elsewhere. While sometimes used for safety (backup systems), in data and logic, it usually indicates inefficiency.
- **In Code:** Writing the same calculation multiple times instead of using a function call.
- **In Data (Normalization):** Storing a customer's address in every single "Order" row instead of once in a "Customer" table.
- **The Goal:** The **"DRY"** principle **(Don't Repeat Yourself)**. Reducing redundancy minimizes the chance of errors when updating information.
3. **Literalness**
> **Literalness** is the quality of being exactly as stated, without any metaphorical, implied, or context-dependent meaning.
- **Computers are Literals:** A computer does exactly what is written in the code, not what the programmer meant to write. If you tell a script to delete a directory, it won't ask "Are you sure?" unless you explicitly coded that check.
- **In Data Types:** A String **"10"** is literally the characters '1' and '0'. Even though it looks like a number to a human, the computer treats it as text until you use a function like **int()** to cast it.
- **The Goal:** Developers must embrace literalness by being extremely precise with syntax, naming conventions, and logic.

## **Debugging**

1. **Types of Bugs**

- **Bugs** are generally categorized by when and how they occur:
    - **Syntax Errors:** Mistakes in the language's "grammar" (e.g., a missing parenthesis or a typo in a function name). These prevent the code from running at all.
    - **Runtime Errors:** Errors that occur while the program is executing (e.g., trying to divide by zero or calling len() on an integer). These cause the program to "crash."
    - **Logic Errors:** The code runs without crashing, but the output is wrong. This happens when the algorithm or mathematical formula is flawed.
    - **Semantic Errors:** The code is technically correct but doesn't do what the programmer intended (often due to ambiguity or a lack of literalness).

2. The Debugging Process

- **Debugging** is a cycle of observation and hypothesis. Most developers follow these steps:
    - **Reproduce:** Find the exact steps or data input that causes the bug to appear consistently.
    - **Isolate:** Narrow down which part of the code (the specific function or line) is responsible.
    - **Analyze:** Use tools to inspect variables and memory at the moment of the error.
    - **Fix:** Apply a correction to the code.
    - **Test:** Verify that the fix works and ensure it didn't create new bugs elsewhere **(Regression Testing)**.

#### **What are the **bitwise** operators in Python?**

Bitwise operators in Python allow you to perform operations on the **individual bits** (0s and 1s) of integers. Instead of treating a number as a single value, Python treats it as a string of binary digits.

Here is the breakdown of the operators and how they function.

---

- **1. The Standard Operators**

| Operator | Name | Description |
| :--- | :--- | :--- |
| `&` | **Binary AND** | Sets each bit to `1` if both bits are `1`. |
| `\|` | **Binary OR** | Sets each bit to `1` if one of two bits is `1`. |
| `^` | **Binary XOR** | Sets each bit to `1` if only one of two bits is `1`. |
| `~` | **Binary NOT** | Inverts all the bits (flips 0 to 1 and 1 to 0). |
| `<<` | **Left Shift** | Shifts bits to the left by pushing zeros in from the right. |
| `>>` | **Right Shift** | Shifts bits to the right by pushing copies of the leftmost bit in from the left. |

---

- **2. How They Work (With Examples)**
    - To understand these, it helps to see the binary representation. Let’s use $a = 10$ (binary `1010`) and $b = 4$ (binary `0100`).

- **AND, OR, and XOR**
    * **AND (`&`)**: `10 & 4` results in `0` (`0000`). Only bits that are `1` in both numbers remain `1`.
    * **OR (`|`)**: `10 | 4` results in `14` (`1110`). If either bit is `1`, the result is `1`.
    * **XOR (`^`)**: `10 ^ 4` results in `14` (`1110`). It returns `1` only if the bits are different.

- **Bitwise NOT (`~`)**
    - The NOT operator is a bit quirky in Python because it uses **Two's Complement** arithmetic. The formula is:
        - $$\sim x = -(x + 1)$$
        - So, `~10` becomes `-11`.

- **Shift Operators**
    * **Left Shift (`<<`)**: Shifting left effectively multiplies a number by $2^n$.
        * `10 << 2` is $10 \times 2^2 = 40$.
    * **Right Shift (`>>`)**: Shifting right effectively performs floor division by $2^n$.
        * `10 >> 2` is $10 // 2^2 = 2$.

---

- **3. Quick Reference Code**
```python
a = 10  # 1010
b = 4   # 0100

print(a & b)  # 0
print(a | b)  # 14
print(a ^ b)  # 14
print(~a)     # -11
print(a << 1) # 20
print(a >> 1) # 5
```

#### **What is the order of operations in Python?** 

In Python, the order of operations follows the standard mathematical convention (PEMDAS/BODMAS), but it expands to include specific programming operators like bitwise logic and comparisons. 

Python evaluates expressions from **highest precedence** to **lowest precedence**. When operators have the same precedence, they are usually evaluated from **left to right** (except for exponentiation).

---

- **1. The Precedence Table**
    - Here is the order of operations, ranked from top (first) to bottom (last):

| Precedence | Operator | Description |
| :--- | :--- | :--- |
| **1** | `( )` | **Parentheses** (Grouped expressions) |
| **2** | `**` | **Exponentiation** (Power) |
| **3** | `+x`, `-x`, `~x` | **Unary** (Positive, Negative, Bitwise NOT) |
| **4** | `*`, `/`, `//`, `%` | **Multiplication, Division, Floor Div, Modulo** |
| **5** | `+`, `-` | **Addition and Subtraction** |
| **6** | `<<`, `>>` | **Bitwise Shifts** |
| **7** | `&` | **Bitwise AND** |
| **8** | `^` | **Bitwise XOR** |
| **9** | `\|` | **Bitwise OR** |
| **10** | `==`, `!=`, `>`, `>=`, etc. | **Comparisons and Identity** (`is`, `in`) |
| **11** | `not` | **Logical NOT** |
| **12** | `and` | **Logical AND** |
| **13** | `or` | **Logical OR** |

---

- **2. Key Rules to Remember**
    - **The Exponentiation Exception**
        - Most operators are evaluated left-to-right. However, **exponentiation (`**`) is right-associative**. This means:
            - $$2 ** 3 ** 2$$
            - is calculated as $2 ** (3 ** 2) = 2^9 = 512$, **not** $(2^3)^2 = 64$.

    - **Comparison Chaining**
        - Python allows you to chain comparisons, which can be tricky. For example, `1 < 2 < 3` is evaluated as `(1 < 2) and (2 < 3)`.

    - **Logical Operators**
        - `and` has higher precedence than `or`. This is a common source of bugs!
```python
# This evaluates 'True and False' first
result = True or True and False 
# Result is True (True or False)
```

---

- **3. Best Practice: Clarity over Cleverness**
    - While knowing the table is useful for interviews or debugging, the most "Pythonic" way to write code is to **use parentheses** whenever an expression might be ambiguous to another human.

> **Bad:** `x = a + b * c >> d`
> 
> **Good:** `x = (a + (b * c)) >> d`

#### **What are the arguments of the round function?**

The `round()` function in Python is used to round a floating-point number to a specified number of decimals. It’s a built-in function, so you don't need to import any modules to use it.

The syntax is:
`round(number, ndigits)`

---

- **The Arguments**

- **1. `number` (Required)**
    - This is the number you want to round. It must be a numeric type (an `int` or a `float`).

- **2. `ndigits` (Optional)**
    - This represents the number of decimal places to which you want to round the number.
        * **If omitted or `None`:** The function returns the nearest **integer**.
        * **If positive:** The number is rounded to that many decimal places.
        * **If negative:** The number is rounded to the left of the decimal point (the nearest 10, 100, etc.).

---

- **Examples in Action**
| Code | Result | Explanation |
| :--- | :--- | :--- |
| `round(3.14159)` | `3` | No `ndigits` provided, rounds to nearest integer. |
| `round(3.14159, 2)` | `3.14` | Rounds to 2 decimal places. |
| `round(123.456, -1)` | `120.0` | Rounds to the nearest 10. |
| `round(123.456, -2)` | `100.0` | Rounds to the nearest 100. |

---

- **Important Note: "Rounding Half to Even"**
    - Python uses a strategy called **Banker's Rounding**. If a number is exactly halfway between two options (e.g., `.5`), Python rounds to the nearest **even** number.

> **Example:**
> * `round(2.5)` becomes `2`
> * `round(3.5)` becomes `4`

This helps reduce cumulative errors in large datasets by preventing a consistent upward bias.

#### **What is the modulus operator in Python?**

The **modulus operator** in Python is represented by the percent symbol (`%`). It returns the **remainder** of a division operation between two numbers.

While the standard division operator (`/`) tells you how many times a number fits into another, the modulus operator tells you what is "left over."

---

- **1. How it Works**
    - The syntax is `a % b`, where `a` is the dividend and `b` is the divisor.


![long-Division](./Images/Long-Division.jpeg) 


* **Expression:** `10 % 3`
* **Logic:** 3 goes into 10 three times ($3 \times 3 = 9$).
* **Remainder:** $10 - 9 = 1$.
* **Result:** `1`

- **Comparison Table**

| Operation | Operator | Result for `10` and `3` | Description |
| :--- | :--- | :--- | :--- |
| **Division** | `/` | `3.3333...` | Standard float division. |
| **Floor Division** | `//` | `3` | How many whole times it fits. |
| **Modulus** | `%` | `1` | The remainder left over. |

---

- **2. Common Use Cases**
    - The modulus operator is incredibly useful in programming for logic that requires "wrapping around" or cycles.

* **Checking for Even/Odd numbers:** If `number % 2 == 0`, the number is even. If it equals `1`, it’s odd.
* **Keeping values in a range:** Useful for game logic (e.g., keeping a player on a grid) or clock math (e.g., `total_hours % 24`).
* **Running code every Nth time:** Inside a loop, `if index % 5 == 0:` will run every 5 iterations.

---

- **3. Modulus with Negative Numbers**
    - Python’s modulus behavior with negative numbers can be surprising if you're coming from C or Java. In Python, the result always takes the **sign of the divisor** ($b$).
    - The formula Python uses is:
        - $$r = a - (b \times (a // b))$$

* `7 % 3` = `1`
* `-7 % 3` = `2` (Because $-7 // 3$ is $-3$, and $-7 - (3 \times -3) = 2$)

---

- **4. Quick Code Example**
```python
print(15 % 4)   # Output: 3
print(10 % 2)   # Output: 0 (Even)
print(11 % 2)   # Output: 1 (Odd)
print(450 % 360) # Output: 90 (Useful for degrees/rotations)
```

- In Python, the behavior of **round()** for a number ending in 0.5 depends on the integer part of that number. Python uses a strategy called **Banker's Rounding** (or "Round Half to Even").
Why does Python do this?

- **Standard rounding** (always rounding 0.5 up) introduces a statistical bias in large datasets. If you have thousands of numbers ending in 0.5 and you always round them up, the average of your resulting dataset will be slightly higher than the average of the original data.

- By rounding to the nearest even number, the "up" and "down" movements tend to cancel each other out, keeping the mean of your data more accurate. This is particularly important in fields like Data Science and Mathematical Optimization.

### **Exercise 4**

The following questions give you a chance to practice writing arithmetic expressions:
- **1. How many seconds are there in 42 minutes 42 seconds?**
- **2. How many miles are there in 10 kilometers? Hint: there are 1.61 kilometers in a mile.**
- **3. If you run a 10 kilometer race in 42 minutes 42 seconds, what is your average pace in seconds per mile?**
- **4. What is your average pace in minutes and seconds per mile?**
- **5. What is your average speed in miles per hour?**

````python
# 1. How many seconds are there in 42 minutes 42 seconds?
seconds = 42 * 60 + 42
print(f"There are {seconds} seconds in 42 minutes 42 seconds")

# 2. How many miles are there in 10 kilometers? 
mile = 10 / 1.61
print(f"There are {round(mile, 2)} miles in 10 Kilometers")

# 3. Average pace in seconds per mile?
# Total seconds divided by total miles
average_seconds_per_mile = seconds / mile
print(f"The average seconds per mile is: {round(average_seconds_per_mile, 2)}")

# 4. Average pace in minutes and seconds per mile?
pace_minutes = int(average_seconds_per_mile // 60)
pace_seconds = int(average_seconds_per_mile % 60)
print(f"Pace: {pace_minutes} minutes and {pace_seconds} seconds per mile")

# 5. What is your average speed in miles per hour?
# Speed = Distance / Time. We need time in hours.
hours = seconds / 3600
speed_mph = mile / hours
print(f"Average speed: {round(speed_mph, 2)} miles per hour")
````
