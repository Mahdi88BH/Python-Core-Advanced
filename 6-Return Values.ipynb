A **pure function** is a fundamental concept in functional programming. At its core, a function is considered "pure" if it behaves like a mathematical operation: it only cares about its input and produces an output without affecting the world around it.

- To be pure, a function must satisfy two strict conditions:
1. **Determinism (Same Input, Same Output)**
- Every time you call the function with the same arguments, it must return the exact same result. It cannot rely on any hidden state, external variables, or random data.
    - **Impure:** A function that uses datetime.now() or random.random().
    - **Pure:** A function that calculates $x + y$.
2. **No Side Effects**
- A pure function does not modify any state outside of its own scope. It is "read-only" regarding the rest of the program.

- **Side effects include:**
    - Modifying a global variable.
    - Changing an attribute of an object passed as an argument.
    - Printing to the console or writing to a file.
    - Triggering a network request.

## **Incremental Development**

**Incremental Development** is a software engineering strategy where you build and test a system in small, manageable pieces rather than trying to write the entire program at once.

> **The philosophy is simple:** "Build a little, test a little." Each small piece you add should result in a functional (even if incomplete) version of the program.

1. **How the Process Works**
- Instead of a "Big Bang" approach where you write 500 lines of code and then hit "Run" for the first time, you follow these steps:
    - **Start with a working foundation:** Write the absolute minimum code to get the program started.
    - **Add one small feature:** Implement just one new function or logic block.
    - **Test immediately:** Run the code. If it breaks, you know exactly where the error is (the 5 lines you just added).
    - **Repeat:** Once the new part works, move on to the next small increment.

```python
def distance(x1, x2, y1, y2):
    return 0.0
```

```python
distance(1, 3, 5, 7)
```

```python
def distance(x1, x2, y1, y2):
    dx = x2 - x1
    dy = y2 - y1
    return dx, dy
```

```python
distance(1, 3, 5, 7)
```

```python
import math

def distance(x1, x2, y1, y2):
    dx = x2 - x1
    dy = y2 - y1
    sum = dx**2 + dy**2
    return math.sqrt(sum)
```

```python
distance(1, 3, 5, 7)
```

## **Leap of Faith**

1. **The Core Concept: The "Leap of Faith**
- "The "Leap of Faith" is the act of assuming a function works perfectly without looking at its internal code.
    - **The Problem:** Following the "flow of execution" (tracing every line) is exhausting. If your code calls a function, which calls another function, which calls another, your brain will quickly run out of "memory."
    - **The Solution:** Treat functions as Black Boxes. You provide the input, you trust the output, and you ignore the "machinery" inside
2. **Why It Works (and why you already do it)**
- The author points out that you are already a "believer" in the leap of faith:
    - **Built-in Functions:** When you use math.sqrt(16), you don't go read the source code for how Python calculates square roots. You just trust it returns 4.0.
    - **Your Own Functions:** Once you have written and tested a function (like is_divisible), you should stop re-reading it. Use it as a reliable tool.
3. **Applying It to Recursion**
- This is where the leap of faith is most important. Recursion is notoriously hard to "trace" because it loops back on itself.
- To write a recursive function (like Factorial), you shouldn't imagine the computer jumping back and forth. Instead, you ask:
    - **The Assumption:** "Assuming my function already knows how to solve the problem for $n - 1$..."
    - **The Logic:** "...how do I use that result to solve for $n$?
    - **"Example (Factorial):**
        - If I trust that factorial(4) returns $24$...
        - Then factorial(5) is simply $5 \times 24$.

| Concept | Description | Why It Matters |
| :--- | :--- | :--- |
| **Flow of Execution** | Tracing every single step and jump a program takes line-by-line. | Useful for debugging small snippets but becomes **overwhelming** in large systems. |
| **The Leap of Faith** | Assuming a function works correctly and returns the right result without looking at its body. | Allows you to focus on the **logic of the current task** rather than internal details. |
| **Black Box Logic** | Treating functions like `math.sqrt()` or your own tested functions as reliable tools. | You don't need to know *how* a tool is made to know *how* to use it effectively. |
| **Recursive Trust** | Assuming the $(n-1)$ case of a recursive call is already solved. | Essential for writing recursion; it turns a complex loop into a simple **mathematical definition**. |
| **Abstraction** | Hiding complexity behind a simple function name. | This is the fundamental building block of **advanced software engineering**. |

## **Fibonacci**

```python
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n-1) + fibonacci(n-2) 
```

```python
fibonacci(3)
```

```python
fibonacci(2)
```

```python
isinstance(3, (int, float)) # equal to isinstance(3, int) isinstance(3, float)
```

```python
def factorial(n):
    if not isinstance(n, int):
        print("factorial in only for integers")
        return None
    elif n < 0:
        print("factorial only for positive (integers) number")
    elif n == 0:
        return 1
    else:
        return n * factorial(n-1)
```

The Ackermann function, A m, n , is defined as:
$$A(m, n) =
\begin{cases} 
n + 1 & \text{if } m = 0 \\
A(m-1, 1) & \text{if } m > 0 \text{ and } n = 0 \\
A(m-1, A(m, n-1)) & \text{if } m > 0 \text{ and } n > 0 
\end{cases}$$
Write a function named ackermann that evaluates the Ackermann function. What happens if you call ackermann(5, 5)?

```python
def Ackermann(m,n):
    if m == 0:
        return n+1
    elif m > 0 and n == 0:
        return Ackermann(m-1, 1)
    elif m > 0 and n > 0:
        return Ackermann(m-1, Ackermann(m, n-1))
    else:
        None
```

```python
Ackermann(3, 4)
```
