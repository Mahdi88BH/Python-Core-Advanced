## **Expression**

- In both mathematics and computer science, an **expression** is a combination of values, variables, and operators that is "evaluated" by the system to produce a single result.

```python
6 * 4 ** 2
```

- The Standard Hierarchy
> When an expression contains multiple operators, they are evaluated in the following sequence:
1. **Parentheses / Brackets:** Anything inside $( )$ or $[ ]$ is solved first
2. **Exponents / Orders:** Powers and square roots (e.g., $2^3$, $\sqrt{9}$).
3. **Multiplication and Division:** These hold equal rank and are processed from left to right.
4. **Addition and Subtraction:** These also hold equal rank and are processed from left to right.

## **Formal and Natural Languages**

1. **Ambiguity**
> **Ambiguity** occurs when a single expression, statement, or command can be interpreted in more than one way.
- **In Language:** "I saw a man with a telescope." (Did I use the telescope, or did the man have it?)
- **In Programming:** If the order of operations isn't clear, an expression like $10 + 5 * 2$ is ambiguous to a human unless they know the language's specific precedence rules.
- **The Goal:** High-quality code and data schemas aim for zero ambiguity so the computer always performs exactly what the developer intended.
2. **Redundancy**
> **Redundancy** refers to the inclusion of information that is unnecessary or already exists elsewhere. While sometimes used for safety (backup systems), in data and logic, it usually indicates inefficiency.
- **In Code:** Writing the same calculation multiple times instead of using a function call.
- **In Data (Normalization):** Storing a customer's address in every single "Order" row instead of once in a "Customer" table.
- **The Goal:** The **"DRY"** principle **(Don't Repeat Yourself)**. Reducing redundancy minimizes the chance of errors when updating information.
3. **Literalness**
> **Literalness** is the quality of being exactly as stated, without any metaphorical, implied, or context-dependent meaning.
- **Computers are Literals:** A computer does exactly what is written in the code, not what the programmer meant to write. If you tell a script to delete a directory, it won't ask "Are you sure?" unless you explicitly coded that check.
- **In Data Types:** A String **"10"** is literally the characters '1' and '0'. Even though it looks like a number to a human, the computer treats it as text until you use a function like **int()** to cast it.
- **The Goal:** Developers must embrace literalness by being extremely precise with syntax, naming conventions, and logic.

## **Debugging**

1. **Types of Bugs**

- **Bugs** are generally categorized by when and how they occur:
    - **Syntax Errors:** Mistakes in the language's "grammar" (e.g., a missing parenthesis or a typo in a function name). These prevent the code from running at all.
    **Runtime Errors:** Errors that occur while the program is executing (e.g., trying to divide by zero or calling len() on an integer). These cause the program to "crash."
    - **Logic Errors:** The code runs without crashing, but the output is wrong. This happens when the algorithm or mathematical formula is flawed.
    - **Semantic Errors:** The code is technically correct but doesn't do what the programmer intended (often due to ambiguity or a lack of literalness).

2. The Debugging Process

- **Debugging** is a cycle of observation and hypothesis. Most developers follow these steps:
    - **Reproduce:** Find the exact steps or data input that causes the bug to appear consistently.
    - **Isolate:** Narrow down which part of the code (the specific function or line) is responsible.
    - **Analyze:** Use tools to inspect variables and memory at the moment of the error.
    - **Fix:** Apply a correction to the code.
    - **Test:** Verify that the fix works and ensure it didn't create new bugs elsewhere (Regression Testing).

- In Python, the behavior of **round()** for a number ending in 0.5 depends on the integer part of that number. Python uses a strategy called **Banker's Rounding** (or "Round Half to Even").
Why does Python do this?

- **Standard rounding** (always rounding 0.5 up) introduces a statistical bias in large datasets. If you have thousands of numbers ending in 0.5 and you always round them up, the average of your resulting dataset will be slightly higher than the average of the original data.

- By rounding to the nearest even number, the "up" and "down" movements tend to cancel each other out, keeping the mean of your data more accurate. This is particularly important in fields like Data Science and Mathematical Optimization.

```python
# 1. How many seconds are there in 42 minutes 42 seconds?
seconds = 42 * 60 + 42
print(f"There are {seconds} seconds in 42 minutes 42 seconds")

# 2. How many miles are there in 10 kilometers? 
mile = 10 / 1.61
print(f"There are {round(mile, 2)} miles in 10 Kilometers")

# 3. Average pace in seconds per mile?
# Total seconds divided by total miles
average_seconds_per_mile = seconds / mile
print(f"The average seconds per mile is: {round(average_seconds_per_mile, 2)}")

# 4. Average pace in minutes and seconds per mile?
pace_minutes = int(average_seconds_per_mile // 60)
pace_seconds = int(average_seconds_per_mile % 60)
print(f"Pace: {pace_minutes} minutes and {pace_seconds} seconds per mile")

# 5. What is your average speed in miles per hour?
# Speed = Distance / Time. We need time in hours.
hours = seconds / 3600
speed_mph = mile / hours
print(f"Average speed: {round(speed_mph, 2)} miles per hour")
```
