## **Defining New Functions**

- In programming, a **function definition** is the actual implementation of a function. It is the block of code that specifies what the function does, how it processes input, and what it returns.

- Think of a function declaration as a "hello" or a promise that a function exists, while the definition is the "how-to" manual.

**Core Components**
- A standard function definition typically includes:
    - **Header:** Specifies the function name, return type, and parameters (inputs).
    - **Body:** The set of statements enclosed in curly braces **{}** (or indentation in languages like Python) that perform the task.
    - **Return Statement:** (Optional) Sends a value back to the part of the program that called the function.

````python
def print_lyrics():
    print("Ana ghadi ldar")
    print("Ana mchit ldar")
````

- In Python, a **function object** is what exists in memory after a function has been defined.

- Unlike many other languages where functions are just "pieces of code," Python treats functions as First-Class Objects. This means a function is an instance of the function class, just like a string is an instance of str or a number is an instance of int.

**What Makes it an "Object"?**
- Because functions are objects, you can treat them like any other data type:
    - **Assignment:** You can assign a function to a variable.
    - **Passing:** You can pass a function as an argument to another function.
    - **Returning:** A function can return another function.
    - **Attributes:** Functions have built-in attributes (like \__name__ or \__doc__).

````python
print_lyrics
````

- that print_lyrics is a function that takes no arguments.
- \__main__ is the name of the module that contains print_lyrics.

- In Python's execution model, a **frame (or frame object)** is a data structure that represents the context in which a function is currently running.

- Whenever you call a function, Python creates a new frame to keep track of that specific execution. If you call the same function three times, Python creates three separate frames.

**What is Inside a Frame?**
- A frame acts like a "container" for a function's execution state. It holds:
    - **Local Variables:** Variables defined inside the function.
    - **The Instruction Pointer:** Keeps track of which line of code is currently being executed.
    - **The Symbol Table:** A map of names to their values (like x = 10).
    - **Global References:** A link back to the global namespace so the function can find variables defined outside of itself.
    - **Return Address:** Where the program should go back to once the function finishes.

**The Call Stack**
- Frames are organized in a Stack (Last-In, First-Out).
    - **Push:** When you call a function, its frame is "pushed" onto the top of the stack.
    - **Active Frame:** Only the frame at the very top of the stack is "active" at any given time.
    - **Pop:** When the function returns a value, its frame is "popped" off the stack and destroyed, and the computer goes back to the frame below it.

**Why Frames Matter**
> **Frames** are the reason recursion works. Each recursive call gets its own frame with its own unique version of local variables, preventing them from overwriting each other.

## **Tracebacks**

- A Traceback is Python’s way of showing you the "history" of a crash. Since we just discussed Frames, you can think of a Traceback as a printed report of the Call Stack at the exact moment an error occurred.

- It shows you every function that was currently active (had a frame on the stack) when the program stopped.

**How to Read a Traceback**
- Python tracebacks are read bottom to top:
1.  **The Bottom Line:** This is the most important part. It tells you the Error Type (e.g., IndexError) and the Error Message (a description of what went wrong).
2. **The "Trace":** Moving upward from the bottom, each section represents a Frame.
    - It shows the File Name and Line Number.
    - It shows the Function Name where the error happened.
    - It shows the specific Line of Code that failed.

**Connecting Frames to Tracebacks**
- When an error happens in a nested function, Python looks at the current Active Frame. If it can't find a way to "handle" the error there (using something like a try/except block), it "pops" that frame and moves to the Previous Frame to see if that one can handle it.

- If it goes all the way down to the Global Frame and still hasn't handled the error, it terminates the program and prints the Traceback.

**Pro Tip: The "Most Recent Call Last"**
- Python explicitly labels its tracebacks with Traceback (most recent call last):. This is your reminder that the code that actually broke is at the very bottom, while the code that started the chain of events is at the top.

## **Why Functions?**

1. **Reusability (DRY Principle)**
- **The most famous rule in programming is DRY:** Don't Repeat Yourself. If you need to calculate the tax on a price in ten different places in your code, you shouldn't write the math formula ten times. If the tax rate changes, you'd have to find and fix all ten spots.
    - **With a Function:** You write the formula once. If the rate changes, you update it in one place (the function definition), and the entire program is fixed.

2. **Abstraction (Hiding Complexity)**
- Abstraction means you can use a tool without needing to understand exactly how its "insides" work.
    - Think of a car’s brake pedal. You don’t need to know about hydraulic fluid or friction coefficients to stop the car; you just "call" the brake.
    - In Python, when you use **print()**, you are calling a function. You don't see the hundreds of lines of code that talk to your computer’s operating system to put text on the screen—you just use the "tool."

3. **Modularity and Organization**
- Functions allow you to break a giant, intimidating problem into "bite-sized" chunks.
- Instead of one block of 1,000 lines, you have 10 functions of 100 lines each. This makes the code scannable. It is much easier to debug calculate_total() than it is to search through a "Wall of Code" for a math error.

4. **Namespace Protection**
- As we discussed with Frames, functions create their own "local" space.
    - If you have a variable named x inside a function, it won't accidentally overwrite a variable named x used somewhere else in your program.
    - This "isolation" prevents bugs where one part of your code accidentally breaks another part.

## **Built-in Function `print()`**

> While we usually just toss a string into print(), the function actually has five parameters. One is for your data, and the four others are "keyword arguments" that control how that data appears on your screen.

1. **\*objects**
- This is the most common parameter. **The asterisk (*)** means it can take any number of objects (strings, numbers, lists, etc.).
    - **Behavior:** It will convert all these objects into strings and print them out, separated by a space.
    - **Example:** print("Hello", 2026, True)

2. **sep (Separator)**
- By default, if you pass multiple objects, Python puts a space between them. The sep parameter lets you change that.
    - **Default:** ' ' (a single space)
    - **Custom Example:** print("Apple", "Banana", sep=", ")
        - **Output:** Apple, Banana

3. **end**
- By default, every time you call **print()**, it adds a newline character (\n) at the end, which moves the cursor to the next line.
    - **Default:** '\n'
    - **Custom Example:** print("Loading", end="...")
    - **Output:** Loading... (the next print statement will start right after the dots instead of on a new line)

4. **file**
- This tells Python where to send the text. Usually, it sends it to the "Standard Output" (your terminal or console).
    - **Default:** sys.stdout
    - **Use Case:** You can redirect the output of a print() statement directly into a text file.
    - **Example:** 
```python
with open("log.txt", "w") as f:
print("Error occurred", file=f)
```

5. **flush**
- Normally, Python "buffers" output—it waits until it has a chunk of data before writing it to the screen to save energy/time. **flush=True** forces Python to clear the buffer and show the text instantly.
    - **Default:** False
    - **Use Case:** Useful for progress bars or real-time logs where you can't afford to wait for the buffer to fill up.

## **Practice**

- In Python, whitespace isn't just for making code look "pretty"—it is a functional part of the language's syntax. While other languages use curly braces {} to define blocks of code, Python uses indentation.

- This makes the choice between Spaces and Tabs one of the most debated topics in the Python community.

**The Golden Rule: Consistency**
- **The absolute most important rule is:** Never mix tabs and spaces in the same file. If you mix them, Python will often throw a TabError: inconsistent use of tabs and spaces in indentation. This happens because, while they might look the same to your eyes, the Python interpreter sees them as different characters with different widths.

Spaces vs. Tabs: The Comparison

| Feature | Spaces (Recommended) | Tabs |
| :--- | :--- | :--- |
| **Standard** | **PEP 8 Choice** (4 spaces) | Discouraged for new projects |
| **Consistency** | Identical across all editors | Width varies by editor settings |
| **Error Risk** | Low | High (Inconsistent use leads to `TabError`) |
| **Visuals** | Represented as dots `....` | Represented as arrows `→` |

**PEP 8: The Official Stance**
- **PEP 8**, the official Style Guide for Python Code, is very clear on this:
    - "Spaces are the preferred indentation method. Tabs should be used solely to remain consistent with code that is already indented with tabs."

**The Industry Standard:** 4 spaces per indentation level.
