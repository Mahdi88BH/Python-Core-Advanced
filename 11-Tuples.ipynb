## **Tuple Assignment**

```python
a, b = 1, 2

print(f'value of a => {a}')
print(f'value of b => {b}')
```

```python
email = 'mahdi@python.org'
username, domain = email.split('@')

username, domain
```

```python
# The number of variables on the left and the number of values on 
# the right have to be the same—otherwise you get a ValueError

# a, b = 1, 2, 3 => ValueError: too many values to unpack (expected 2)
```

```python
a, b = b, a

print(f'value of a => {a}')
print(f'value of b => {b}')
```

- This works because all of the expressions on the right side are evaluated before any of the assignments.

```python
# We can also use tuple assignment in a for statement.

dict_value = {'a': 1, 'b': 2}

for item in dict_value.items():
    key, value = item
    print(f'key => {key} and value => {value}')
```

## **Argument Packing**

- Functions can take a variable number of arguments. A parameter name that begins with the **\* operator packs** arguments into a tuple.

```python
def mean(*args):
    return sum(args) / len(args)
```

```python
mean(1, 2, 3, 4, 5)
```

- If you have a sequence of values and you want to pass them to a function as multiple arguments, you can use the * operator to **unpack** the tuple.

```python
t = (5, 3)
divmod(*t)
```

```python
t = [5, 3]
divmod(*t)
```

```python
def min_max(t):
    return min(t), max(t)

def mean(*args):
    return sum(args) / len(args)

def trimmed_mean(*args):
    min, max = min_max(args)
    trimmed = list(args)
    trimmed.remove(min)
    trimmed.remove(max)

    return mean(*trimmed)
```

```python
min_max([1, 2, 3, 10])
```

```python
trimmed_mean(1, 2, 3, 10)
```

## **Comparing and Sorting**

- The relational operators work with tuples and other sequences. For example, if you use the **< operator** with tuples, it starts by comparing the first element from each sequence. If they are equal, it goes on to the next pair of elements, and so on, until it finds a pair that differ

```python
(0, 1, 200000) < (0, 3, 4)
```

```python
def value_counts(word):
    letters_dict = {}
    for item in word:
        if item not in letters_dict:
            letters_dict[item] = 1
        else:
            letters_dict[item] += 1
    
    return letters_dict
```

```python
items = value_counts('banana').items()

items
```

```python
def value_counts2(word):
    letters_dict = {}
    for item in word:
        if item not in letters_dict:
            letters_dict[item] = 1
        else:
            letters_dict[item] += 1
    # dict_items object that behaves like a list of tuples
    return sorted(letters_dict.items()) 
```

```python
value_counts2('banana')
```

```python
def second_element(t):
    return t[1]

def value_counts3(word):
    letter_dict = {}
    for letter in word:
        if letter not in letter_dict:
            letter_dict[letter] = 1
        else:
            letter_dict[letter] += 1
    
    return sorted(letter_dict.items(), key=second_element, reverse=True)
```

## **Inverting a Dictionary**

```python
def letter_counter(word):
    counter = {}
    for letter in word:
        if letter in counter:
            counter[letter] += 1
        else:
            counter[letter] = 1
    return counter
```

```python
letter_counter('parrot')
```

```python
def invert_dict(word):
    counter = letter_counter(word)
    new_dict = {}

    for key, value in counter.items():
        if value not in new_dict:
            new_dict[value] = [key]
        else:
            new_dict[value].append(key)
    
    return new_dict
        
```

```python
invert_dict('parrot')
```

## **Glossary**
- **tuple:** An immutable object that contains a sequence of values.
- **pack:** Collect multiple arguments into a tuple.
- **unpack:** Treat a tuple (or other sequence) as multiple arguments.
- **zip object:** The result of calling the built-in function zip, can be used to loop through a sequence of tuples.
- **enumerate object:** The result of calling the built-in function enumerate, can be used to loop through a sequence of tuples.
- **sort key:** A value, or function that computes a value, used to sort the elements of a collection.
- **data structure:** A collection of values, organized to perform certain operations efficiently.
