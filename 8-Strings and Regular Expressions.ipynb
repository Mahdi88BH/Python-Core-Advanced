## **String Comparison**

```python
word = 'Banana'
if word == 'banana':
    print('All right, bananas.')
else:
    print('I do not like bananas.')
```

```python
def compare_word(word):
    if word > 'banana':
        print(word, 'come after banana')
    elif word < 'banana':
        print(word, 'come before banana')
    else:
        print('The Are equals')
```

```python
compare_word('elmahdi')
```

```python
compare_word('Elmahdi')
```

> **Important Rule:** Case Sensitivity Python treats **Uppercase letters** as "smaller" than Lowercase letters because their **Unicode values** are lower.

## **Regular Expressions**

```python
text = "I am Dracula; and I bid you welcome, Mr. Harker, to my house."

text
```

```python
pattern = 'Dracula'

pattern
```

```python
import re

result = re.search(pattern, text)

print(f"the text that was searched => {result.string}")
print(f"the part of the text that matched the pattern => {result.group()}")
print(f"returns the index in the text where the pattern starts and ends {result.span()}")
```

- If the **pattern** doesn’t appear in the text, the return value from search is None

```python
result = re.search('banana', text)

print(f"The text that was searched => {result}")
```

```python
result == None
```

```python
def find_first(pattern):
    for line in open('./Dracule_replaced_text.txt'):
        result = re.search(pattern, line)
        if result != None:
            return result.string
```

```python
result = find_first('Elmahdi')

result
```

```python
pattern = r'Mina|Murray'
result = find_first(pattern)

result
```

```python
def count_match(pattern):
    count = 0
    for line in open('./Dracule_replaced_text.txt', 'r'):
        if re.search(pattern, line) != None:
            count += 1
    return count
```

```python
print(f'There are {count_match("Elmahdi")} occurrences of "Elmahdi" in the text.')
```

```python
print(f'There are {count_match("Mina|Murray")} occurrences of "Mina|Murray" in the text.')
```

```python
# Mathe The beginning of string
result = find_first('^Dracula')

result
```

```python
# Mathe The ending of string
result = find_first('Harker$')

result
```

## **String Substitution**

```python
# this pattern matches a sequence that starts with 'cent' and ends with 
# either 'er' or 're'
pattern = "cent(er|re)"
result = find_first(pattern)

result
```

```python
# The following pattern uses the special character '?', which means that 
# the previous character is optional
pattern = "colou?r"
result = find_first(pattern)

result
```

```python
modified_result = re.sub(pattern, 'color', result)

modified_result
```

- **re.search()** returns a Match Object. This object contains metadata, like the original string (.string) and where the match was found.
- **re.sub()** returns a String. It performs the replacement immediately and gives you the resulting text.

## **Glossary**
- **sequence:** An ordered collection of values where each value is identified by an integer index.
- **character:** An element of a string, including letters, numbers, and symbols.
- **index:** An integer value used to select an item in a sequence, such as a character in a string. In Python, indices start from 0.
- **slice:** A part of a string specified by a range of indices.
- **empty string:** A string that contains no characters and has length 0.
- **object:** Something a variable can refer to. An object has a type and a value.
- **immutable:** If the elements of an object cannot be changed, the object is immutable.
- **invocation:** An expression—or part of an expression—that calls a method.
- **regular expression:** A sequence of characters that defines a search pattern.
- **pattern:** A rule that specifies the requirements a string has to meet to constitute a match.
- **string substitution:** Replacement of a string, or part of a string, with another string.
- **shell command:** A statement in a shell language, which is a language used to interact with an operating system.
- **raw string:** A Python string that is preceded by the letter r, which indicates that backslashes that appear in the string should not be considered part of a special sequence.
