## **Creating a Point**

````python
from copy import copy
````

````python
class Point:
    """Represents a point in 2-D space"""

    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __str__(self):
        return f'Points: ({self.x}, {self.y})'
    
    def __eq__(self, other):
        return (self.x == other.x) and (self.y == other.y)
    
    def translate(self, dx, dy):
        self.x += dx
        self.y += dy

    def translated(self, dx=0, dy=0):
        point = copy(self)

        point.x += dx
        point.y += dy

        return point
````

````python
start = Point(0, 0)

print(start)
````

````python
end = start.translated(0, 150)

print(end)
````

## **Creating a Line**

````python
from jupyturtle import make_turtle, jumpto, moveto
````

````python
class Line:
    """Represent a Line with Coordinates x-axis, y-axis"""

    def __init__(self, p1, p2):
        self.p1 = p1
        self.p2 = p2
    
    def __str__(self):
        return f'Coordinates are: (({self.p1}), ({self.p2}))'
    
        

    def draw(self):
        jumpto(self.p1.x, self.p1.y)
        moveto(self.p2.x, self.p2.y)
````

````python
line = Line(start, end)

print(line)
````

````python
line.draw()
````

## **Creating a Rectangle**

````python
class Rectangle:
    """
    Represents a rectangle.
    attributes: width, height, corner.
    """

    def __init__(self, width, height, corner):
        self.width = width
        self.height = height
        self.corner = corner

    def __str__(self):
        return f'dimension of Rectangle:\n- width=> {self.width}\n- height=> {self.height}\n- at corner {self.corner}'

    def make_point(self):
        p1 = self.corner
        p2 = p1.translated(self.width, 0)
        p3 = p1.translated(0, self.height)
        p4 = p1.translated(-self.width, 0)

        return p1, p2, p3, p4
    
    def make_lines(self):
        p1, p2, p3, p4 = self.make_point()
        return Line(p1, p2), Line(p2, p3), Line(p3, p4), Line(p4, p1)
    
    def draw_line(self):
        lines = self.make_lines()
        for line in lines:
            line.draw()

    def grow(self, dx, dy):
        box = copy(self)

        box.width += dx
        box.height += dy

        return box
    
    def translate(self, dx, dy):
        self.corner.translate(dx, dy)


````

## **Deep Copy**

- When we use **copy** to duplicate box1, it copies the **Rectangle object** but not the **Point object** it contains. So box1 and box2 are different objects, as intended

````python
box1 is box2
````

````python
box1.corner is box2.corner
````

- The following diagram shows the state of these objects:

![Rectangle_state_diagrame](./Images/Rectangle_state_diagrame.png)

- What copy does is create a **shallow copy** because it copies the object but **not the objects it contains**. As a result, changing the width or height of one Rectangle does not affect the other, but changing the attributes of the **shared Point** affects both! This behavior is confusing and error prone.
- Fortunately, the copy module provides another function, called **deepcopy**, that copies not only the object but also the objects it refers to, and the objects they refer to, and so on. This operation is called a **deep copy**.

````python
corner = Point(20, 20)
box3 = Rectangle(100, 50, corner)

print(box3)
````

````python
from copy import deepcopy
````

````python
box4 = deepcopy(box3)

print(box4)
````

````python
box3.corner is box4.corner
````

````python
box3.translate(50, 30)
box4 = box4.grow(100, 60)
````

````python
print(box3, box4)
````

````python
box3.corner is box4.corner
````

## **Debugging**

This chapter is all about **Shared Mutable State**—the number one source of "subtle bugs" in object-oriented programming.

The problem occurs when two different objects hold a reference to the **same** underlying object. If you change the underlying object through one path, the other object changes too, often unexpectedly.

---

- **The Core Problem: Aliasing**
    - Imagine you have two `Rectangle` objects, `box1` and `box2`. If they both share the *same* `Point` object as their corner, modifying `box1` will secretly modify `box2` as well. This is called **aliasing**, and it is a silent killer in complex programs.

---

- **Two Strategies for Bug Prevention**
    - To avoid these "spooky" side effects, the chapter suggests two main strategies:

- **1. Avoid Sharing: Deep Copy**
    - If you absolutely must have two objects that *look* the same but remain independent, you must use a **Deep Copy**.
* **What it does:** It creates a brand-new object in memory and recursively copies every attribute inside it.
* **Result:** The new object is a completely separate entity. Modifying it will not affect the original.

---

- **2. Avoid Modifying: Use Pure Functions**
    - Instead of creating "modifiers" (functions that change an object in place), use "pure functions" that return a new object.

* **Modifier (The "Danger" approach):**
```python
  def translate(self, dx, dy):
      self.x += dx  # Modifies existing object!
      self.y += dy
```

* **Pure Function (The "Safe" approach):**
```python
  def translated(self, dx, dy):
      # Creates and returns a BRAND NEW Point
      return Point(self.x + dx, self.y + dy)
```

---

- **Summary Checklist**

| Strategy | Goal | How it works |
| :--- | :--- | :--- |
| **Deep Copy** | Prevent Sharing | You get an independent clone of the original object. |
| **Pure Function** | Prevent Mutation | You get a new object with new values, leaving the original untouched. |

- **Why this matters**
    - By following these rules, your code becomes **predictable**. You stop having to worry, *"If I change this object, what else in my program is going to break?"* Instead, you gain confidence that your objects are stable, independent, and safe to use throughout your application.

## **Glossary**

- **identical:** Being the same object (which implies equivalence).
- **equivalent:** Having the same value.
- **shallow copy:** A copy operation that does not copy nested objects.
- **deep copy:** A copy operation that also copies nested objects.
- **polymorphism:** The ability of a method or operator to work with multiple types of objects.

#### **How do I make a Python object immutable?**

Making an object immutable in Python means ensuring its internal state cannot be modified after it has been created. Because Python does not have a native "immutable" keyword for classes, you must use specific design patterns to enforce this behavior.

---

- **1. Using `@dataclass(frozen=True)` (The Modern Way)**
    - This is the cleanest and most standard way to create immutable objects in modern Python. By setting `frozen=True`, Python will raise an `AttributeError` if you try to modify any attribute after the object is initialized.

```python
from dataclasses import dataclass

@dataclass(frozen=True)
class ImmutablePoint:
    x: int
    y: int

p = ImmutablePoint(10, 20)
# p.x = 30  # This will raise an AttributeError!
```

---

- **2. Overriding `__setattr__` (The Custom Way)**
    - If you are not using data classes, you can manually override the magic method `__setattr__`. This method is called whenever an attribute assignment is attempted; by forcing it to raise an error, you effectively lock the object.

```python
class LockedPoint:
    def __init__(self, x, y):
        super().__setattr__('x', x)
        super().__setattr__('y', y)

    def __setattr__(self, name, value):
        raise AttributeError("This object is immutable")
```

---

- **3. Using `@property` (The "Read-Only" Way)**
    - You can define attributes as private (using a leading underscore) and expose them only through "getter" methods using the `@property` decorator. This prevents external code from overwriting the data directly.

```python
class ReadOnlyPoint:
    def __init__(self, x, y):
        self._x = x
        self._y = y
    
    @property
    def x(self):
        return self._x

# Now, p.x = 10 will fail, even though the internal _x exists.
```

---

- **Important Considerations: Deep Immutability**
    - A common pitfall is **shallow vs. deep immutability**. If your class contains a mutable object (like a `list` or a `dict`), even if your class is "frozen," the **content** of that list can still be changed.

- To be truly immutable:
    * **Avoid mutable containers:** Use `tuples` instead of `lists`.
    * **Use `frozenset`:** Instead of `set`.
    * **Return copies:** In your methods, always return a new instance of the object rather than modifying the current one.

---

- **Summary Checklist**

| Strategy | Best Used For |
| :--- | :--- |
| **`frozen=True`** | Clean, modern class definitions (Recommended). |
| **`__setattr__`** | Custom objects requiring strict locking logic. |
| **`@property`** | When you need calculated properties or strict encapsulation. |

````python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    
    def __str__(self):
        return f'coordinate of point: ({self.x}, {self.y})'

    def __eq__(self, other):
        return (self.x == other.x) and (self.y == other.y)
````

````python
class Line:
    def __init__(self, p1, p2):
        self.p1 = p1
        self.p2 = p2

    def __str__(self):
        return f'coordinate of line are : ({self.p1}, {self.p2})'
    
    def __eq__(self, other):
        return (self.p1 == other.p1) and (self.p2 == other.p2)
    
    def midpoint(self):
        return Point((self.p1.x + self.p2.x)/2, (self.p1.y + self.p2.y)/2)
````

````python
class Rectagle:
    def __init__(self, width, height, corner):
        self.width = width
        self.height = height
        self.corner = corner
    
    def midpoint(self):
        corner_x = self.corner.x + (self.width / 2)
        corner_y = self.corner.y + (self.height / 2)

        return Point(corner_x, corner_y)
````
