## **Parents and Children**

````python
class Hand(Decks):
    """Represents a hand of playing cards."""
````

````python
def __init__(self, lable=''):
    self.lable = lable
    self.cards = []

Hand.__init__ = __init__
````

````python
hand = Hand('player 1')

print(hand.lable)
````

````python
# deck = Decks([card])
card = deck.take_card()
hand.add_card(card)

print(hand)
````

- When a new class **inherits** from an existing one, the existing one is called the **parent** and the new class is called the **child**. In general:
    - Instances of the **child** class should have all of the **attributes** of the **parent** class, but they can have **additional attributes**.
    - The **child** class should have all of the **methods** of the **parent** class, but it can have **additional methods**.
    - If a **child** class **overrides a method** from the parent class, the new method should take **the same parameters** and return a compatible result.

> This set of rules is called the **“Liskov substitution principle”** after computer scientist Barbara Liskov.

````python
def move_card(self, other, num):
    for i in range(num):
        card = self.take_card()
        other.add_card(card)

Decks.move_card = move_card
````

> This method is **polymorphic** that is, it works with more than one type: self and other can be either a Hand or a Deck. So we can use this method to deal a card from Deck to a Hand, from one Hand to another, or from a Hand back to a Deck.

## **Specialization**

````python
class BridgeHand(Hand):
    """Represents a bridge hand."""
    
    hcp_dict = {
    'Ace': 4,
    'King': 3,
    'Queen': 2,
    'Jack': 1,
    }
````

````python
rank = 12
rank_name = Card.rank_names[rank]
score = BridgeHand.hcp_dict.get(rank_name, 0)

rank_name, score
````

````python
def high_card_point_count(self):
    count = 0
    for card in self.cards:
        rank_name = Card.rank_names[card.rank]
        count += BridgeHand.hcp_dict.get(rank_name, 0)
    return count

BridgeHand.high_card_point_count = high_card_point_count
````

````python
hand = BridgeHand('player 2')

deck.shuffle()
deck.move_card(hand, 5)

print(hand)
````

````python
hand.high_card_point_count()
````

- **BridgeHand** inherits the variables and methods of Hand and adds a class variable and a method that are specific to bridge. This way of using inheritance is called **specialization** because it defines a new class that is **specialized for a particular use**, like playing bridge.

## **Debugging**

This chapter addresses the double-edged sword of **Inheritance**. While it is a powerful tool for code reuse, it introduces complexity that can make debugging difficult because the code for a single object is often scattered across multiple parent classes.

---

- **1. The Trade-off of Inheritance**

* **The Pros:**
    * **Conciseness:** You don't have to rewrite the same methods for every new class.
    * **Code Reuse:** You can create "specialized" versions of a class (subclasses) while keeping the original "general" class (parent) untouched.
    * **Natural Modeling:** It maps well to real-world concepts (e.g., a `BridgeHand` is a type of `Hand`, which contains a `Deck`).

* **The Cons:**
    * **Readability:** When you call `object.method()`, it is not immediately obvious *which* class in the inheritance chain actually defines that method. You have to mentally trace up the tree.

---

- **2. Tracing the "Hidden" Flow**
    - When inheritance makes your program flow hard to follow, the chapter suggests two ways to regain control:

- **A. The Simple Trace (Print Statements)**
    - Don't overcomplicate it. Add a `print()` statement at the start of your methods.
```python
def shuffle(self):
    print("Running Deck.shuffle")
    # ... rest of the code
```
This gives you a clear log of exactly what is happening in the order it happens.

- **B. The Automated Detective (`find_defining_class`)**
    - If you have a complex class hierarchy, you can use the `find_defining_class` function provided in the chapter. It uses Python’s **Method Resolution Order (MRO)** to act like a search engine for your code.

* **What is MRO?** It is the specific search path Python follows to find a method. Python checks the child class first, then moves up to the parents one by one until it finds the definition.
* **How `find_defining_class` works:** It iterates through the MRO list and asks, "Do you have this method?" as soon as it gets a "Yes," it tells you exactly which class owns that method.
```python
def find_defining_class(obj, method_name):
    """
    """
    for typ in type(obj).mro():
        if method_name in vars(typ):
            return typ
    return f'Method {method_name} not found.'
```

---

- **3. Key Concepts Checklist**

| Concept | Definition |
| :--- | :--- |
| **Inheritance** | Defining a new class based on an existing one to share code. |
| **MRO** | **Method Resolution Order**: The list of classes Python searches to find a method implementation. |
| **Tracing** | The act of following the execution path (using prints or tools) to debug inheritance issues. |

- **Summary**
    - Inheritance is great for building structures, but it hides the "where" of your code. When you find yourself lost in a web of parent and child classes, use **`type(obj).mro()`** or a helper function like **`find_defining_class`** to see the reality behind the abstraction.

#### **What are some pros and cons of object-oriented programming?**

**Object-Oriented Programming (OOP)** is a paradigm structured around the concept of **objects**, which contain both data (attributes) and code (methods). While it is one of the most widely used paradigms in software engineering, it comes with a specific set of trade-offs.

---

- **The Pros: Why use OOP?**
    * **Modularity and Reusability:** Through **inheritance**, you can create a base class and reuse its logic in multiple subclasses. This prevents code duplication and makes it easier to organize large codebases.
    * **Encapsulation:** By hiding the internal state of an object (private attributes) and only exposing necessary methods, you create a "black box." This protects the object from accidental modification and makes it easier to change internal logic without breaking the rest of your system.
    * **Easier Maintenance:** Because related data and behaviors are grouped together, it is often easier to locate and fix bugs. If something is wrong with a `User` object, you know exactly where the code for that user lives.
    * **Polymorphism:** This allows you to treat different objects as if they were the same type (e.g., a `Circle` and a `Square` can both respond to a `.draw()` method). This makes your code more flexible and easier to extend.

---

- **The Cons: Where OOP struggles**
    * **Increased Complexity:** Simple programs can become bloated in an OOP structure. You might find yourself writing "boilerplate" code (getters, setters, class constructors) just to accomplish a simple task that could be done with a single function.
    * **Steep Learning Curve:** Understanding concepts like abstraction, polymorphism, and inheritance, and knowing when to use them correctly, can take significant time for beginners.
    * **Memory and Performance Overhead:** Because objects often involve complex structures and metadata, they can be more memory-intensive than simple data structures (like arrays or primitive types).
    * **"The Banana-Monkey Problem":** As the programmer Joe Armstrong famously noted, *"You wanted a banana but what you got was a gorilla holding the banana and the entire jungle."* Because of deep inheritance hierarchies, you often have to instantiate a massive object just to access a tiny piece of functionality.

---

- **Comparison Table**

| Feature | OOP Approach | Alternative (Procedural/Functional) |
| :--- | :--- | :--- |
| **Logic** | Data and behavior are bundled. | Data and logic are separated. |
| **Expansion** | Extensible via inheritance. | Extensible via function composition. |
| **State** | Managed internally (mutable). | Managed externally (often immutable). |
| **Complexity** | High upfront, lower at scale. | Low upfront, can become messy at scale. |

---

- **When should you use it?**
    - OOP is generally **excellent for large, complex systems** where you need to manage multiple entities that share behavior (like a game, a UI framework, or a banking application). However, if you are writing small scripts, data processing pipelines, or high-performance mathematical functions, a more functional or procedural approach might be more efficient and easier to read.

Most modern software is a **hybrid**: you use OOP to manage the structure and state of your application, but use functional techniques for the actual data processing logic inside your methods.

#### **What does it mean when people say ‘favor composition over inheritance’?**

"Favor composition over inheritance" is a core principle in object-oriented design that suggests you should build complex objects by **combining smaller, independent objects** (composition) rather than by creating deep, rigid hierarchies of parent and child classes (inheritance).

In simpler terms: **Build by "having" rather than "being."**

---

- **1. The Inheritance Approach ("Is-A")**
    - Inheritance creates a rigid relationship. If you have a `Bird` class and create a `Penguin` subclass, the penguin "is a" bird. If the `Bird` class has a `fly()` method, the penguin inherits it—even though penguins can't fly. You then have to "hack" the code to prevent the penguin from flying.
        * **The Problem:** Your objects become tightly coupled. If you change the base class, you might accidentally break everything that inherits from it.

- **2. The Composition Approach ("Has-A")**
    - Composition avoids this by building objects out of smaller parts. Instead of a `Bird` class that *does* everything, you create a `Flying` behavior and a `Swimming` behavior as separate components.
    - A `Penguin` is just an object that *has* a `Swimming` component but *lacks* a `Flying` component.

---

- **The Key Differences**

| Feature | Inheritance | Composition |
| :--- | :--- | :--- |
| **Relationship** | "Is-A" (Penguin *is a* Bird) | "Has-A" (Penguin *has a* Swimming ability) |
| **Coupling** | Tight (child depends on parent) | Loose (objects are independent) |
| **Flexibility** | Rigid (hierarchy is set at compile time) | Highly flexible (behaviors can be swapped) |
| **Design** | Top-down (designing big trees) | Bottom-up (assembling building blocks) |

---

- **Why "Favor" Composition?**
    1.  **Avoids the "Fragile Base Class" Problem:** When you use inheritance, a small change in a high-level parent class can ripple down and break dozens of subclasses. Composition keeps changes localized.
    2.  **Increased Flexibility:** With composition, you can change an object’s behavior at runtime. For example, if a `Robot` object has a `Movement` component, you can swap it from `Walking` to `Wheeled` while the program is running. You cannot do this with inheritance.
    3.  **Easier Testing:** It is much easier to test a small, focused "behavior" component in isolation than to test a massive class that has inherited thousands of lines of code from a long chain of ancestors.

---

- **The Practical Takeaway**
    - You don't need to abandon inheritance entirely—it is useful for creating specialized versions of a base object. However, before you create a deep class tree, ask yourself:
        * *"Am I creating this hierarchy because the objects truly share an 'Is-A' relationship?"*
        * *"Or am I just trying to share some code functionality?"*

> If you are just trying to share functionality, **composition** is almost always the safer, more maintainable choice.

#### **What is the Liskov substitution principle?**

The **Liskov Substitution Principle (LSP)** is the "L" in the SOLID principles of object-oriented design. It states that:

> **Objects of a superclass should be replaceable with objects of its subclasses without breaking the application.**

In simpler terms: If you have a program that works with a base class, it should continue to work perfectly if you swap that base class instance with an instance of any of its subclasses.

---

- **1. The Core Idea**
    - When you use inheritance, your subclasses should adhere to the "contract" defined by the parent class. If a subclass changes the expected behavior (e.g., throwing an error where the parent didn't, or returning a different type of data), it violates the Liskov principle.

---

- **2. A Classic Example: The Rectangle-Square Problem**
    - Imagine you have a `Rectangle` class with a method `set_width` and `set_height`.

```python
class Rectangle:
    def set_width(self, width): self.width = width
    def set_height(self, height): self.height = height

class Square(Rectangle):
    # A square forces width and height to be the same
    def set_width(self, width):
        self.width = width
        self.height = width
```

> **This violates LSP. Why?** Because a piece of code that expects a `Rectangle` will assume that `set_width` only changes the width. If you pass it a `Square`, calling `set_width` *also* changes the height. Your program now behaves in an unexpected way. The `Square` is not a true substitute for a `Rectangle`.

---

- **3. How to avoid violating LSP**
    * **Don't force inheritance:** If you find yourself adding `if` statements or `try-except` blocks to check if an object is a specific subclass before using it, you are likely violating LSP.
    * **Respect the Contract:** Subclasses should fulfill the same "promises" as the parent class. If the parent method is expected to return a positive integer, the subclass should not return a negative one.
    * **Favor Interfaces/Abstract Classes:** Instead of inheriting deep implementation details, have your classes implement a shared **interface** that clearly defines what behaviors are required.

---

- **4. Summary Table**

| Principle Aspect | Impact |
| :--- | :--- |
| **Substitution** | You should be able to swap base for derived classes seamlessly. |
| **Contract** | Subclasses must not break the logic established by the parent. |
| **Design** | Encourages smaller, more focused interfaces. |

- **Why does it matter?**
    - If your code violates LSP, it becomes fragile. You end up with "if-this-is-type-X" logic scattered everywhere to handle the "special" behavior of specific subclasses. By following LSP, you ensure that your code is **extensible**—you can add new subclasses without ever having to modify the code that uses the parent class.

#### **Is Python an object-oriented language?**

In fact, one of the core design philosophies of Python is that **"everything is an object."** This means that not only are custom classes and instances objects, but basic data types like integers, strings, functions, and even modules are objects.

---

- **Why Python is considered Object-Oriented**

1.  **Everything is an Object:** In Python, every variable you create is an instance of a class. When you type `x = 5`, Python creates an instance of the `int` class. You can even see this for yourself by checking `type(x)` in your code.

--- 

2.  **Supports OOP Pillars:** Python provides full support for the four fundamental pillars of Object-Oriented Programming:
    * **Encapsulation:** You can hide internal data using naming conventions (like `_variable` or `__variable`) and manage access via properties and methods.
    * **Inheritance:** You can create classes that derive behavior from parent classes, enabling code reuse and hierarchical organization.
    * **Polymorphism:** Different classes can define the same method name (e.g., `.area()`), and Python will execute the correct one based on the object type.
    * **Abstraction:** You can use Abstract Base Classes (via the `abc` module) to define blueprints that force subclasses to implement specific methods.

---

3.  **Flexibility (Multi-Paradigm):** While Python is object-oriented, it is also a **multi-paradigm language**. This means it does not *force* you to use classes if you don't want to. You can write:
    * **Procedural code:** Using simple functions and variables.
    * **Functional code:** Using concepts like `map`, `filter`, `lambda` functions, and list comprehensions.
    * **Object-Oriented code:** Using classes and inheritance.

---

- **The Benefit of this Design**
    - Because Python treats everything as an object, you get a consistent interface for everything. Whether you are working with a list, a custom class, or an integer, you have access to methods and attributes in the same way (using dot notation). 

> However, because it is multi-paradigm, Python is often more approachable for beginners than strictly object-oriented languages like Java, where you *must* create a class and a `main` method just to print "Hello World."

#### **What are the requirements for a set to be totally ordered?**

A set $S$ is **totally ordered** (also called a linear order) if it possesses a **partial order** that satisfies one additional, critical condition.

To be a totally ordered set, the relation (often denoted by $\le$) must satisfy four specific requirements.

---

- **1. The Four Requirements**
    - For a set $S$ and a relation $\le$ to form a total order, the following must be true for all elements $a, b, c \in S$:
        1.  **Reflexivity:** Every element is related to itself.
            * $a \le a$
        2.  **Antisymmetry:** If two elements are related to each other in both directions, they must be the same element.
            * If $a \le b$ and $b \le a$, then $a = b$.
        3.  **Transitivity:** The order is consistent across a chain of elements.
            * If $a \le b$ and $b \le c$, then $a \le c$.
        4.  **Comparability (Totality):** This is the defining feature that makes a *partial* order *total*. Every pair of elements must be comparable.
            * Either $a \le b$ or $b \le a$ (or both).

---

- 2. **Understanding the Difference**
    * **Partial Order:** Requires Reflexivity, Antisymmetry, and Transitivity. Some pairs of elements might be "incomparable." 
        * *Example:* The "subset" relation ($\subseteq$) on sets. If you have set $A=\{1\}$ and set $B=\{2\}$, neither $A \subseteq B$ nor $B \subseteq A$. They are incomparable.
    * **Total Order:** Adds the **Comparability** requirement. There are no "incomparable" pairs. You can always line the elements up in a specific, unbroken sequence.

---

- **3. Concrete Examples**

| Set | Relation | Totally Ordered? | Why? |
| :--- | :--- | :--- | :--- |
| **Integers ($\mathbb{Z}$)** | $\le$ | **Yes** | For any two integers, one is always less than or equal to the other. |
| **Real Numbers ($\mathbb{R}$)** | $\le$ | **Yes** | The number line is a continuous, linear sequence. |
| **Power Set $\mathcal{P}(S)$** | $\subseteq$ | **No** | As shown above, subsets can be incomparable. |
| **Complex Numbers ($\mathbb{C}$)** | $\le$ | **No** | There is no standard way to say if $1+i$ is "less than" $2+3i$. |

---

- **Why does "Total" Order matter?**
- The utility of a total order is that it allows for **sorting**. If you have a set that is only partially ordered, you cannot arrange the elements in a single list from "smallest" to "largest." In computer science, most sorting algorithms (like Quicksort or Mergesort) require the underlying set to be totally ordered to function correctly.
