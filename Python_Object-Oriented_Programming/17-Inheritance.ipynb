## **Representing Cards**

````python
class Card:
    """Represents a standard playing card."""
    
    suit_names = ['Clubs', 'Diamonds', 'Hearts', 'Spades']
    rank_names = [None, 'Ace', '2', '3', '4', '5', '6', '7',
    '8', '9', '10', 'Jack', 'Queen', 'King', 'Ace']
````

````python
Card.suit_names[0]
````

````python
Card.rank_names[11]
````

## **Card Attributes**

````python
def __init__(self, suit, rank):
    self.suit = suit
    self.rank = rank

Card.__init__ = __init__ 
````

````python
queen = Card(1, 12)
````

````python
queen.suit, queen.rank
````

- It is also legal to use the instance to access the **class variables**

````python
queen.suit_names
````

## **Printing Cards**

````python
def __str__(self):
    suit_name = Card.suit_names[self.suit]
    rank_name = Card.rank_names[self.rank]
    
    return f'{rank_name} of {suit_name}'

Card.__str__ = __str__
````

````python
print(queen)
````

- The following is a diagram of the **Card class object** and the Card instance. Card is a **class object**, so its type is type. queen is an instance of Card, so its type is **Card**:

![diagram](./Images/Daigram_Card_Class_Object.png)

> Every **Card instance** has its own suit and rank attributes, but there is only one Card **class object**, and only one copy of the **class variables** suit_names and rank_names.

## **Comparing Cards**

````python
queen2 = Card(1, 12)

print(queen)
print(queen2)
````

````python
queen == queen2
````

````python
def __eq__(self, other):
    return (self.suit == other.suit) and (self.rank == other.rank)

Card.__eq__ = __eq__
````

````python
queen == queen2
````

````python
six = Card(1, 6)

print(six)
````

````python
six == queen
````

- If we use the **!=** operator, Python invokes a special method called **`__ne__`**, if it exists. Otherwise, it invokes **`__eq__`** and inverts the result so if **`__eq__`** returns True, the result of the != operator is False

````python
queen != queen2
````

````python
# queen < queen2 => TypeError: '<' not supported between instances of 'Card' and 'Card'
````

````python
def to_tuple(self):
    return self.suit, self.rank

Card.to_tuple = to_tuple
````

````python
def __lt__(self, other):
    return self.to_tuple() < other.to_tuple()

Card.__lt__ = __lt__
````

````python
six < queen
````

- If we use the **> operator**, it invokes a special method called **`__gt__`**, if it exists. Otherwise it invokes **`__lt__`** with the arguments in the opposite order

````python
queen > queen2
````

- inally, if we use the **<= operator**, it invokes a special method called **`__le__`**:

````python
def __le__(self, other):
    return self.to_tuple() <= other.to_tuple()


Card.__le__ = __le__
````

````python
queen <= queen2
````

````python
queen >= queen2
````

````python
queen <= six
````

- If we use the **>= operator**, it uses **`__ge__`** if it exists. Otherwise, it invokes **`__le__`** with the arguments in the opposite order:

````python
queen >= six
````

````python
six >= queen2
````

## **Decks**

````python
class Decks:

    def __init__(self, cards):
        self.cards = cards
````

````python
def make_cards():
    cards = []
    for i in range(4):
        for j in range(2, 15):
            card = Card(i, j)
            cards.append(card)
    
    return cards

Decks.make_cards = make_cards
````

````python
cards = Decks.make_cards()
deck = Decks(cards)

len(deck.cards)
````

## **Printing the Deck**

````python
def __str__(self):
    res = []
    for card in self.cards:
        res.append(str(card))
    return '\n'.join(res)

Decks.__str__ = __str__
````

````python
print(deck)
````

````python
small_deck = Decks([queen, six])

print(small_deck)
````

## **Add, Remove, Shuffle, and Sort**

````python
def take_card(self):
    return self.cards.pop()

Decks.take_card = take_card
````

````python
poped_cards = deck.take_card()

print(poped_cards)
````

````python
len(deck.cards)
````

````python
def add_card(self, card):
    self.cards.append(card)

Decks.add_card = add_card
````

````python
deck.add_card(poped_cards)

len(deck.cards)
````

````python
import random
````

````python
def shuffle(self):
    random.shuffle(self.cards)

Decks.shuffle = shuffle
````

````python
deck.shuffle()

for card in deck.cards[:5]:
    print(card)
````

````python
def sort_card(self):
    self.cards.sort()

Decks.sort_card = sort_card
````

- When we invoke sort, it uses the **`__lt__`** method to compare cards:

````python
deck.sort_card()
````

````python
for card in deck.cards[:5]:
    print(card)
````

> In this example, **Deck.sort** doesn’t do anything other than invoke **list.sort**. Passing along responsibility like this is called **delegation**.

## **Parents and Children**

````python
class Hand(Decks):
    """Represents a hand of playing cards."""
````

````python
def __init__(self, lable=''):
    self.lable = lable
    self.cards = []

Hand.__init__ = __init__
````

````python
hand = Hand('player 1')

print(hand.lable)
````

````python
# deck = Decks([card])
card = deck.take_card()
hand.add_card(card)

print(hand)
````

- When a new class **inherits** from an existing one, the existing one is called the **parent** and the new class is called the **child**. In general:
    - Instances of the **child** class should have all of the **attributes** of the **parent** class, but they can have **additional attributes**.
    - The **child** class should have all of the **methods** of the **parent** class, but it can have **additional methods**.
    - If a **child** class **overrides a method** from the parent class, the new method should take **the same parameters** and return a compatible result.

> This set of rules is called the **“Liskov substitution principle”** after computer scientist Barbara Liskov.

````python
def move_card(self, other, num):
    for i in range(num):
        card = self.take_card()
        other.add_card(card)

Decks.move_card = move_card
````

> This method is **polymorphic** that is, it works with more than one type: self and other can be either a Hand or a Deck. So we can use this method to deal a card from Deck to a Hand, from one Hand to another, or from a Hand back to a Deck.

## **Specialization**

````python
class BridgeHand(Hand):
    """Represents a bridge hand."""
    
    hcp_dict = {
    'Ace': 4,
    'King': 3,
    'Queen': 2,
    'Jack': 1,
    }
````

````python
rank = 12
rank_name = Card.rank_names[rank]
score = BridgeHand.hcp_dict.get(rank_name, 0)

rank_name, score
````

````python
def high_card_point_count(self):
    count = 0
    for card in self.cards:
        rank_name = Card.rank_names[card.rank]
        count += BridgeHand.hcp_dict.get(rank_name, 0)
    return count

BridgeHand.high_card_point_count = high_card_point_count
````

````python
hand = BridgeHand('player 2')

deck.shuffle()
deck.move_card(hand, 5)

print(hand)
````

````python
hand.high_card_point_count()
````

- **BridgeHand** inherits the variables and methods of Hand and adds a class variable and a method that are specific to bridge. This way of using inheritance is called **specialization** because it defines a new class that is **specialized for a particular use**, like playing bridge.

## **Debugging**

This chapter addresses the double-edged sword of **Inheritance**. While it is a powerful tool for code reuse, it introduces complexity that can make debugging difficult because the code for a single object is often scattered across multiple parent classes.

---

- **1. The Trade-off of Inheritance**

* **The Pros:**
    * **Conciseness:** You don't have to rewrite the same methods for every new class.
    * **Code Reuse:** You can create "specialized" versions of a class (subclasses) while keeping the original "general" class (parent) untouched.
    * **Natural Modeling:** It maps well to real-world concepts (e.g., a `BridgeHand` is a type of `Hand`, which contains a `Deck`).

* **The Cons:**
    * **Readability:** When you call `object.method()`, it is not immediately obvious *which* class in the inheritance chain actually defines that method. You have to mentally trace up the tree.

---

- **2. Tracing the "Hidden" Flow**
    - When inheritance makes your program flow hard to follow, the chapter suggests two ways to regain control:

- **A. The Simple Trace (Print Statements)**
    - Don't overcomplicate it. Add a `print()` statement at the start of your methods.
```python
def shuffle(self):
    print("Running Deck.shuffle")
    # ... rest of the code
```
This gives you a clear log of exactly what is happening in the order it happens.

- **B. The Automated Detective (`find_defining_class`)**
    - If you have a complex class hierarchy, you can use the `find_defining_class` function provided in the chapter. It uses Python’s **Method Resolution Order (MRO)** to act like a search engine for your code.

* **What is MRO?** It is the specific search path Python follows to find a method. Python checks the child class first, then moves up to the parents one by one until it finds the definition.
* **How `find_defining_class` works:** It iterates through the MRO list and asks, "Do you have this method?" as soon as it gets a "Yes," it tells you exactly which class owns that method.
```python
def find_defining_class(obj, method_name):
    """
    """
    for typ in type(obj).mro():
        if method_name in vars(typ):
            return typ
    return f'Method {method_name} not found.'
```

---

- **3. Key Concepts Checklist**

| Concept | Definition |
| :--- | :--- |
| **Inheritance** | Defining a new class based on an existing one to share code. |
| **MRO** | **Method Resolution Order**: The list of classes Python searches to find a method implementation. |
| **Tracing** | The act of following the execution path (using prints or tools) to debug inheritance issues. |

- **Summary**
    - Inheritance is great for building structures, but it hides the "where" of your code. When you find yourself lost in a web of parent and child classes, use **`type(obj).mro()`** or a helper function like **`find_defining_class`** to see the reality behind the abstraction.

## **Glossary**

- **inheritance:** The ability to define a new class that is a modified version of a previously defined class.
- **encode:** To represent one set of values using another set of values by constructing a mapping between them.
- **class variable:** A variable defined inside a class definition, but not inside any method.
- **totally ordered:** A set of objects is totally ordered if we can compare any two elements and the results are consistent.
- **delegation:** When one method passes responsibility to another method to do most or all of the work.
- **parent class:** A class that is inherited from.
- **child class:** A class that inherits from another class.
- **specialization:** A way of using inheritance to create a new class that is a specialized version of an existing class.
