## **Creating Python classes**

```python
class MyFirstClass:
    pass
```

```python
a = MyFirstClass()
b = MyFirstClass()

print(a)
print(b)
```

### **Adding attributes**

```python
class Point:
    pass
```

```python
p1 = Point()
p2 = Point()

p1.x = 2
p1.y = 5


p2.x = 4
p2.y = 8
```

```python
print("P1 :", p1.x, "," ,p1.y)
print("P2 :", p2.x, "," ,p2.y)
```

## **Making it do something**

```python
class Point:
    def reset(self):
        self.x = 0
        self.y = 0
```

```python
p = Point()
p.reset()
print(f"point P: {p.x}, {p.y}")
```
