### **Talking to yourself**

> Pay attention to the difference between a **class** and an **object** in this discussion. We can think of the method as a function attached to a class. The self parameter is a specific instance of that class. When you call the method on two different objects, you are calling the same method twice, but passing two different objects as the self parameter.

````python
p1 = Point()
Point.reset(p1)

print(f"P1: {p1.x}, {p1.y}")
````

In Python, the `self` argument represents the **specific instance of the class** that is currently executing a method.

It is the mechanism that allows Python to know *which* object’s data it should be reading or modifying. Without `self`, a class would have no way to distinguish between multiple individual objects created from the same blueprint.

---

- **🚗 The Target Analogy**
    - Imagine you write a single class called `Car` that has a method called `paint(color)`. If you instantiate two separate objects—`car_a` and `car_b`—and then call `paint("Red")`, how does Python know which car to turn red?
    - **`self`** acts as the pointer. When you call a method on `car_a`, Python automatically passes `car_a` into the `self` slot behind the scenes. It's the equivalent of saying: *"Take **this specific car** and change its color."*

---

- **💻 How It Works in Code**
    - When defining a method inside a class, you must explicitly include `self` as the **very first parameter**. However, when you actually *call* the method on an object, you do not pass an argument for `self`; Python injects it automatically.

```python
class Robot:
    def __init__(self, name):
        # 'self.name' tells Python to attach the data to THIS specific robot instance
        self.name = name 

    def introduce(self):
        # 'self' allows the method to look up data belonging to this instance
        print(f"🤖 Hello, my name is {self.name}.")

# 1. Create two independent objects
wall_e = Robot("Wall-E")
eva = Robot("Eve")

# 2. Call the method
wall_e.introduce()  # Output: 🤖 Hello, my name is Wall-E.
eva.introduce()     # Output: 🤖 Hello, my name is Eve.
```

- **What Python is doing under the hood:**
    - When you write `wall_e.introduce()`, Python secretly translates that line into this:

```python
# Python passes the specific object instance explicitly into the class function
Robot.introduce(wall_e)
```

Because `wall_e` is passed in, `self.name` resolves to `wall_e.name`, producing "Wall-E".

---

- **🔍 Why do we have to explicitly type `self`?**
    - If you've used languages like Java, C++, or C#, you might be used to the **`this`** keyword, which is implicit (you don't have to declare it in the method signature). Python's creator, Guido van Rossum, chose to make **`self`** explicit based on the core Zen of Python design philosophy: **"Explicit is better than implicit."**
    - Declaring **`self`** makes it entirely transparent that the method is an instance method designed to interact with an object's state, rather than a detached utility function or class-level method.

---

- **⚖️ Common Gotchas to Avoid**
    * **Forgetting `self` in the definition:** If you define a method like `def speak():`, Python will throw a `TypeError: speak() takes 0 positional arguments but 1 was given` the moment you try to call it on an object, because Python tried to force-feed the object instance into it.
    * **Forgetting `self` when accessing variables:** Inside a method, writing `print(name)` will cause Python to search for a local variable or a global variable. To read an object's attribute, you must write `print(self.name)`.
    * **The name isn't legally fixed:** The word `self` isn't a strict keyword in Python; it is just a strong community convention. You could technically name it `this` or `banana`, and the code would run perfectly. However, using anything other than `self` will make your code incredibly confusing to any other Python developer.

## **Initializing the object**

In object-oriented programming, the terms **constructor** and **initializer** are often used interchangeably, but under the hood, they represent two completely distinct phases of an object's birth: **the constructor creates the object, while the initializer populates it with data.**

Python handles this distinction explicitly by separating the process into two distinct magic methods: **`__new__`** (The Constructor) and **`__init__`** (The Initializer).

---

- **🏗️ 1. The Constructor (`__new__`)**
    - The **constructor** is the actual architect. Its sole job is to allocate raw physical space in your computer's memory (RAM) and build the empty shell of the object.
        * **When it runs:** It is the absolute first method called when you type `Object()`.
        * **What it returns:** It **must** return a brand new, blank instance of the class.
        * **Python Method:** `__new__(cls, *args, kwargs)`

---

- **🎨 2. The Initialization (`__init__`)**
    - The **initializer** is the interior designer. Once the constructor finishes building the empty object shell in memory, the initializer steps inside to configure its starting state, assign initial data values to its attributes, and run any setup logic.
        * **When it runs:** It automatically triggers immediately *after* the constructor has successfully created and returned the new object.
        * **What it returns:** It **cannot** return anything (it implicitly returns `None`). Its only job is to modify the `self` object that was passed to it by the constructor.
        * **Python Method:** `__init__(self, *args, kwargs)`

---

- **💻 Visualizing the Split in Python Code**
    - While 99% of the time you will only ever need to write an `__init__` method in your daily Python code, Python allows you to override `__new__` to see exactly how the baton is passed between the two:

```python
class Spacerocket:
    def __new__(cls, name):
        print("1. 🏗️ __new__ (Constructor) runs FIRST: Allocating RAM memory space...")
        # Super creates the actual physical, blank instance of the object
        instance = super().__new__(cls)
        return instance # Passes this blank shell forward to __init__

    def __init__(self, name):
        print("2. 🎨 __init__ (Initializer) runs SECOND: Binding data attributes...")
        # Customizing the shell that __new__ built
        self.name = name

# --- The Birth of an Object ---
rocket = Spacerocket("Falcon Heavy")
```

---

- **📊 Side-by-Side Comparison**

| Feature | Constructor (`__new__`) | Initializer (`__init__`) |
| --- | --- | --- |
| **Primary Goal** | **Object Creation.** Allocates memory for the instance. | **State Setup.** Assigns initial values to variables. |
| **First Argument** | Receives the class itself (`cls`). | Receives the newly created instance (`self`). |
| **Return Value** | Must return a newly created object instance. | Must not return anything (Returns `None`). |
| **Core Metaphor** | Manufacturing an empty smartphone factory-sealed box. | Unboxing the phone, setting the wallpaper, and logging into Wi-Fi. |

````python
class Point:
    
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def move(self, x, y):
        self.x = x
        self.y = y

    def reset(self):
        self.move(0, 0)
````

````python
point = Point(3, 4)
print(f"Points Cordinate's : {point.x}, {point.y}")
point.move(5, 5)
print(f"Points Cordinate's : {point.x}, {point.y}")
point.reset()
print(f"Points Cordinate's : {point.x}, {point.y}")
````

## **Explaining yourself**

````python
import math
````

````python
class Point:
    """Represents a point in two-dimensional geometric coordinates"""

    def __init__(self, x, y):
        """Initialize the position of a new point. The x and y
        coordinates can be specified. If they are not, the
        point defaults to the origin."""
        
        self.x = x
        self.y = y

    def move(self, x, y):
        """Move the point to a new location in 2D space."""
        
        self.x = x
        self.y = y

    def reset(self):
        """Reset the point back to the geometric origin: 0, 0"""
        
        self.move(0, 0)

    def calc_distance(self, other):
        """Calculate the distance from this point to a second
        point passed as a parameter.
        
        This function uses the Pythagorean Theorem to calculate
        the distance between the two points. The distance is
        returned as a float."""
        
        return math.sqrt(
            (self.x - other.x) ** 2
            + (self.y - other.y) **2
        )
````

````python
point = Point(3, 4)
````

````python
point.reset()
````

````python
point.move(3, 5)
````

## **Modules and packages**

While **modules** and **packages** are both mechanisms used in Python to organize, scale, and modularize your code, the difference comes down to their structural complexity: **a module is a single file, whereas a package is a collection of modules inside a directory structure.**

---

- **📄 1. The Module (A Single File)**
    - A **module** is simply a single file ending in the `.py` extension. It contains executable code, variable definitions, functions, and classes that are grouped logically together.
        * **The Metaphor:** Think of a module like a single **chapter** inside a textbook. It focuses thoroughly on a single topic.
        * **How you use it:** You import the file using its filename (without the `.py`).

- **🛠️ Example Structure:**
    - If you create a file named `calculator.py`:

```python
# calculator.py (This is a module)
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b
```

You can seamlessly use this module in another file by writing: `import calculator`.

---

- **📦 2. The Package (A Directory of Modules)**
    - A **package** is a directory (folder) that contains multiple modules grouped together. It allows you to create a hierarchical, nested structure for massive codebases so you don't end up with hundreds of disorganized individual files floating around your root directory.
        * **The Metaphor:** Think of a package like the entire **textbook**. It is a single bound container that holds many different chapters (modules) inside it.
        * **The Structural Rule:** In older versions of Python (Python 2 and early Python 3), a folder *must* contain a file named `__init__.py` to be recognized by the system as a package. In modern Python, this file is technically optional (known as a Namespace Package), but it is still heavily used to initialize package-level configurations or cleanly expose APIs.

- **🛠️ Example Structure:**

```text
my_game_app/            <-- This is a Package (A Directory)
│
├── __init__.py         <-- Makes this directory an official package
├── graphics.py         <-- This is a module inside the package
├── audio.py            <-- This is a module inside the package
└── physics.py          <-- This is a module inside the package
```

- To use a specific capability out of this package, you navigate down the directory chain using dot notation:

```python
from my_game_app import audio
audio.play_laser_sound()
```

---

- **📊 Side-by-Side Comparison**

| Feature | Module | Package |
| --- | --- | --- |
| **Physical Form** | A single `.py` file. | A file directory (folder). |
| **Contents** | Classes, functions, variables, and executable code. | Multiple modules, sub-packages, and an `__init__.py` file. |
| **Import Syntax** | `import module_name` | `from package_name import module_name` |
| **Complexity Level** | Small to medium scale (designed for single-topic logic). | Large scale (designed to manage enterprise frameworks or library structures). |

### **Organizing modules**

#### **Absolute vs. Relative Imports**

When your Python project grows beyond a single file, you need to import modules from different directories. Python handles this using two distinct approaches: **absolute imports** and **relative imports**.

The fundamental difference lies in their starting point: **absolute imports search from the project's root directory down**, while **relative imports search from the current file's position out.**

To illustrate the differences, assume we are working inside the following project structure:

```text
my_project/
│
├── main.py
└── src/
    ├── __init__.py
    ├── core.py
    └── utils/
        ├── __init__.py
        ├── formatters.py
        └── validators.py
```

---

- **🌎 1. Absolute Imports (The Full GPS Address)**
    - An **absolute import** specifies the complete path to the target module, starting all the way from the project's root directory or the system's `sys.path`. It uses the full dot-separated directory path to pinpoint the file.
    - If you are working inside `src/utils/validators.py` and you want to import a function from `src/core.py`, an absolute import looks like this:

```python
# Inside src/utils/validators.py
from src.core import engine_setup
```

- **🟢 Advantages**
    * **Clear and Readable:** It is instantly obvious exactly which file is being imported, no matter how deep you are inside the directory structure.
    * **Stable Movement:** You can copy and paste this import line into a completely different file anywhere else in your project, and it will still work perfectly without modification.
    * **PEP 8 Preferred:** The official Python style guide explicitly states that absolute imports are preferred because they avoid confusing routing bugs.

- **🔴 Disadvantages**
    * **Brittle to Renaming:** If you decide to rename your top-level folder from `src` to `app`, you have to hunt down and change every single absolute import statement across your entire codebase.

---

- **📍 2. Relative Imports (The Walking Directions)**
    - A **relative import** specifies the path to the target module relative to the location of the *current file*. It uses a **dot (`.`) syntax** to navigate up and down through the local package layout:
        * A single dot (`.`) means "look inside the current folder."
        * A double dot (`..`) means "go up one folder level."
        * A triple dot (`...`) means "go up two folder levels," and so on.

Using our same scenario, if you are inside `src/utils/validators.py` and want to import something from its neighbor `src/utils/formatters.py`, or move up to `src/core.py`:

```python
# Inside src/utils/validators.py

# Neighbor import (same directory)
from . import formatters  

# Moving up one level to look inside src/
from ..core import engine_setup  
```

- **🟢 Advantages**
    * **Highly Portable Packages:** If you are building a reusable library or plugin, relative imports allow the entire package to be renamed or moved inside another project without breaking a single internal import.
    * **Shorter Code:** Avoids typing out long, redundant root path strings for deeply nested sub-packages.

- **🔴 Disadvantages**
    * **The "Main Script" Crash:** Relative imports rely strictly on the `__name__` attribute of the module. If you try to run a file containing relative imports directly as a standalone script (e.g., typing `python src/utils/validators.py` in your terminal), Python will crash with a notorious error:
-> `ImportError: attempted relative import with no known parent package`
    * **Harder to Refactor:** If you drag-and-drop a file into a different folder, all of its relative dot paths break instantly and must be re-calculated manually.

---

- **📊 Side-by-Side Comparison**

| Feature | Absolute Imports | Relative Imports |
| --- | --- | --- |
| **Starting Point** | Project root directory (`sys.path`). | Location of the current executing file. |
| **Syntax Style** | `from project_name.module import function` | `from ..module import function` |
| **PEP 8 Status** | **Highly Recommended** (Default standard). | Accepted only for specialized local packages. |
| **Standalone Execution** | Works perfectly when running files directly. | Destroys runtime execution if executed directly. |
| **Best Used For** | Main applications, scripts, and general development. | Standalone shared libraries and deep nested sub-packages. |
