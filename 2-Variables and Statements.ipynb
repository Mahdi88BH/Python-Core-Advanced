## **Variable Names**

> In Python, variable names (also called identifiers) must follow a specific set of rules to ensure the interpreter can understand your code. While the language is flexible, these rules are strictly enforced.

1. **Mandatory Rules (Syntax)**
- If you break these rules, Python will raise a SyntaxError and your code will not run.
    - **Allowed Characters:** Variable names can only contain letters (a-z, A-Z), digits (0-9), and underscores (_).
    - **Start Rule:** A name must start with a letter or an underscore. It cannot start with a digit.
        - **user_1** is valid.
        - **1_user** is invalid.
    - **Case Sensitivity:** Python is case-sensitive. price, Price, and PRICE are three completely different variables.
    - **No Reserved Keywords:** You cannot use words that Python already uses for its own logic (e.g., if, else, while, def, class, True, False).

2. **Standard Conventions (Best Practices)**
- While these aren't "laws," following them makes your code professional and readable for others.
    - **Snake Case:** In Python, the standard for variables is snake_case. Use all lowercase letters and separate words with underscores.
        - **Example:** house_price_prediction, total_seconds.
    - **Meaningful Names:** Avoid single-letter names like x or y unless they are for simple math coordinates or loop counters. Use names that describe the data.
        - Instead of **a = 10**, use **kilometers = 10**.
    - **Constants:** If a variable is meant to stay the same throughout the entire program (a constant), use all capital letters.
        - Example: PI = 3.14159, MAX_ITERATIONS = 100.

3. **Special Use of Underscores**
- Underscores have specific meanings in the Python community:
    - **Single Leading Underscore (_variable):** A hint to other programmers that the variable is intended for "internal use" within a class or module.
    - **Double Leading Underscore (__variable):** Triggers **"name mangling"** to help avoid naming conflicts in classes (more advanced).
    - **Double Leading and Trailing (__init__):** Reserved for special **"dunder"** (double-underscore) methods that have built-in meaning in Python. Never create your own names using this format.

## **Expressions and Statements**

1. **Expressions**
> An expression is a combination of values, variables, operators, and function calls that the Python interpreter evaluates to create a single value.
- **Characteristics:** It can always be reduced to a single result (like a number or a string).
- Examples:
    - 2 + 2 (evaluates to 4)
    - len("Data") (evaluates to 4)
    - miles * 1.61 (evaluates to a float)

2. **Statements**
> A statement is a complete unit of execution that represents an action or a command. It tells the computer to do something.
- **Characteristics:** Statements often contain expressions, but they do not "resolve" to a value themselves. You cannot print a statement's result because it doesn't have one; it just happens.
- Examples:
    - **Assignment Statement:** x = 10 (Action: Store 10 in x)
    - **Import Statement:** import math (Action: Load the math module)
    - **Print Statement:** print("Hello") (Action: Display text on screen)

````python
# This is an Expression
expression = 4 * math.sqrt(round(25.3)) / 2

expression
````

> Computing the value of an expression is called **evaluation**. Running a statement is called **execution**.

> Using **int**, **float**, and **str** as variable names is considered a "bad practice" in Python because these names are built-in functions. When you use them as variable names, you are performing what is known as Shadowing.

Here is why shadowing these names can break your code and cause confusion:

1. **You "Break" the Built-in Function**
- In Python, names are just labels pointing to objects. By default, the name int points to the function that converts things into integers. If you write int = 10, the name int now points to the number 10.
- If you try to use the actual function later in your script, Python will throw an error because you can't "call" a number like a function.
    - **The Bug Scenario:**
```python
int = 10  # You just "shadowed" the built-in function
age = "25"
user_age = int(age) # ERROR: 'int' object is not callable
```

2. **Loss of Literalness and Increased Ambiguity**
- As a data science student, your code needs to be literal and precise. Using str = "Mahdi" creates ambiguity.
- When another developer (or you, three months from now) reads the code, they might expect str to be the function, not a specific piece of data.
- It makes debugging much harder because the error messages (like TypeError: 'str' object is not callable) can be very confusing if you don't realize you've redefined a core part of the language.

3. **Syntax Highlighting Conflicts**
- Most code editors (like VS Code on your Linux setup) highlight built-in functions in a specific color (e.g., blue or purple).
- If you use them as variables, the editor will still highlight them as functions, but your code will treat them as data.
- This visual inconsistency makes it harder to spot syntax errors and logic flaws at a glance.

4. **Collaborative and Library Issues**
- If you are working on a project with a colleague or using shared repositories (like your Comptoire-DWH project), shadowing built-ins can cause "hidden bugs." If your colleague tries to use float() to process a new data point but you've redefined it earlier in the file, their part of the code will crash.

- **Part 1.** The volume of a sphere with radius r is 4/3*πr^3. What is the volume of a sphere with radius 5? Start with a variable named radius and then assign the result to a variable named volume. Display the result. Add comments to indicate that radius is in centimeters and volume is in cubic centimeters.

````python
import math

raduis = 5 # raduis in centimeters
volume = 4/3 * math.pi * math.pow(raduis, 3) # Volume is in cubic centimeters

print("The volume of a sphere with raduis = 5 is =>", round(volume, 3))
````

- **Part 2.** A rule of trigonometry says that for any value of x, **(cos(x))^2 + (sin(x))^2 = 1**. Let’s see if it’s true for a specific value of x like 42.

````python
x = 42
result = math.pow(math.cos(x), 2) + math.pow(math.sin(x), 2)

print("the result of (cos(42))^2 + (sin(42))^2 is =>", int(result))
````

2. Use math.pow to raise math.e to the power 2.
