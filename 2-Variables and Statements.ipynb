## **Variable Names**

> In Python, variable names (also called identifiers) must follow a specific set of rules to ensure the interpreter can understand your code. While the language is flexible, these rules are strictly enforced.

1. **Mandatory Rules (Syntax)**
- If you break these rules, Python will raise a SyntaxError and your code will not run.
    - **Allowed Characters:** Variable names can only contain letters (a-z, A-Z), digits (0-9), and underscores (_).
    - **Start Rule:** A name must start with a letter or an underscore. It cannot start with a digit.
        - **user_1** is valid.
        - **1_user** is invalid.
    - **Case Sensitivity:** Python is case-sensitive. price, Price, and PRICE are three completely different variables.
    - **No Reserved Keywords:** You cannot use words that Python already uses for its own logic (e.g., if, else, while, def, class, True, False).

2. **Standard Conventions (Best Practices)**
- While these aren't "laws," following them makes your code professional and readable for others.
    - **Snake Case:** In Python, the standard for variables is snake_case. Use all lowercase letters and separate words with underscores.
        - **Example:** house_price_prediction, total_seconds.
    - **Meaningful Names:** Avoid single-letter names like x or y unless they are for simple math coordinates or loop counters. Use names that describe the data.
        - Instead of **a = 10**, use **kilometers = 10**.
    - **Constants:** If a variable is meant to stay the same throughout the entire program (a constant), use all capital letters.
        - Example: PI = 3.14159, MAX_ITERATIONS = 100.

3. **Special Use of Underscores**
- Underscores have specific meanings in the Python community:
- Underscores in Python aren't just for making names readable; they act as "status symbols" for variables and methods. Here is a deeper look at how they function with code examples.

---

* **1. Single Leading Underscore (`_variable`)**
    - In Python, there is no such thing as a truly "private" variable. A single underscore is a **"gentleman's agreement"** or a social hint. It tells other developers: *"Hey, I'm using this internally for this class. Don't touch it from the outside, or things might break."*

- **Example:**
```python
class DigitalVault:
    def __init__(self):
        self.public_name = "Guest"
        self._internal_key = "ABC-123"  # Hint: Private!

vault = DigitalVault()
print(vault.public_name)   # Totally fine
print(vault._internal_key) # Python allows this, but it's considered "bad form"
```

**Note:** If you use `from module import *`, variables starting with `_` will not be imported.

---

* **2. Double Leading Underscore (`__variable`)**
    - This is more than just a hint; it triggers **Name Mangling**. Python physically changes the name of the variable to make it harder to accidentally overwrite in a subclass. This is useful when you want to ensure a variable stays unique to its specific class.

- **Example:**
```python
class Parent:
    def __init__(self):
        self.__secret = "Parent Password"

class Child(Parent):
    def __init__(self):
        super().__init__()
        self.__secret = "Child Password"

p = Parent()
# Trying to print(p.__secret) would throw an AttributeError!
# Python renamed it behind the scenes to: _Parent__secret
print(p._Parent__secret) 
```


---

* **3. Double Leading and Trailing (`__init__`)**
    - These are called **Dunder Methods** (short for **D**ouble **Under**score). These are "magic" methods that Python calls automatically under certain conditions.

* **`__init__`**: Called when you create a new object.
* **`__str__`**: Called when you use `print(object)`.
* **`__add__`**: Called when you use the `+` operator on objects.

- **Example:**
```python
class Book:
    def __init__(self, title):
        self.title = title

    def __str__(self):
        return f"Book titled: {self.title}"

my_book = Book("Python Pro")
print(my_book)  # This triggers __str__ automatically!
```
**Rule of thumb:** Only use these exact names when you want to hook into Python's built-in behavior. Never invent your own (e.g., `__my_cool_method__`), as Python might add a method with that name in a future update, breaking your code.

---

* **Summary Comparison**

| Format | Name | Purpose |
| :--- | :--- | :--- |
| `_var` | Internal | A "keep out" sign for humans; no functional change. |
| `__var` | Mangled | Automatically renames the variable to avoid naming collisions. |
| `__var__` | Dunder | Special system methods; reserved for Python's core logic. |

## **Expressions and Statements**

1. **Expressions**
> An expression is a combination of values, variables, operators, and function calls that the Python interpreter evaluates to create a single value.
- **Characteristics:** It can always be reduced to a single result (like a number or a string).
- Examples:
    - 2 + 2 (evaluates to 4)
    - len("Data") (evaluates to 4)
    - miles * 1.61 (evaluates to a float)

2. **Statements**
> A statement is a complete unit of execution that represents an action or a command. It tells the computer to do something.
- **Characteristics:** Statements often contain expressions, but they do not "resolve" to a value themselves. You cannot print a statement's result because it doesn't have one; it just happens.
- Examples:
    - **Assignment Statement:** x = 10 (Action: Store 10 in x)
    - **Import Statement:** import math (Action: Load the math module)
    - **Print Statement:** print("Hello") (Action: Display text on screen)

Here is why shadowing these names can break your code and cause confusion:

1. **You "Break" the Built-in Function**
- In Python, names are just labels pointing to objects. By default, the name int points to the function that converts things into integers. If you write int = 10, the name int now points to the number 10.
- If you try to use the actual function later in your script, Python will throw an error because you can't "call" a number like a function.
    - **The Bug Scenario:**
```python

```

2. **Loss of Literalness and Increased Ambiguity**
- As a data science student, your code needs to be literal and precise. Using str = "Mahdi" creates ambiguity.
- When another developer (or you, three months from now) reads the code, they might expect str to be the function, not a specific piece of data.
- It makes debugging much harder because the error messages (like TypeError: 'str' object is not callable) can be very confusing if you don't realize you've redefined a core part of the language.

3. **Syntax Highlighting Conflicts**
- Most code editors (like VS Code on your Linux setup) highlight built-in functions in a specific color (e.g., blue or purple).
- If you use them as variables, the editor will still highlight them as functions, but your code will treat them as data.
- This visual inconsistency makes it harder to spot syntax errors and logic flaws at a glance.

## **Debugging**

Errors are a natural part of coding—even for the pros. In Python, errors generally fall into three distinct buckets based on **when** they happen and **why** they happen.

---

### **1. Syntax Errors**
**The "Grammar" Error.**
This happens when you break the rules of the Python language. Python can’t even begin to run your code because it doesn't understand your "sentences."

* **Example:** Forgetting a colon at the end of an `if` statement or mismatching parentheses.
* **The Code:**
    ```python
    if True
        print("Hello") # Error: Missing colon after True
    ```
* **How to Debug:**
    1.  **Look at the arrow:** Python usually points a `^` symbol at the exact spot it got confused.
    2.  **Check the line above:** Sometimes the error is actually on the line right before the one Python flagged (like a missing closing parenthesis).
    3.  **Use an IDE:** Most editors (like VS Code or PyCharm) will put a red squiggly line under syntax errors before you even hit "Run."

---

#### **2. Runtime Errors (Exceptions)**
**The "Crash" Error.**
The syntax is perfect, and the program starts running, but it hits a "dead end" it can’t recover from—like trying to do something mathematically impossible or accessing something that doesn't exist.

* **Example:** Dividing by zero or calling a variable that hasn't been defined yet.
* **The Code:**
    ```python
    x = 10
    print(x / 0) # ZeroDivisionError: division by zero
    ```
* **How to Debug:**
    1.  **Read the Traceback:** Python provides a "Traceback," which is a map of where the crash happened. Start at the **bottom** of the error message to see the error type (e.g., `NameError`, `TypeError`).
    2.  **Add Print Statements:** Insert `print()` calls before the crash to see what your variables actually contain right before the error occurs.
    3.  **Try-Except Blocks:** If you expect a specific error might happen (like a user entering a string instead of a number), wrap the code in a `try/except` block to handle it gracefully.

---

#### **3. Semantic Errors**
**The "Logic" Error.**
These are the trickiest because **the program runs perfectly without crashing**, but the result is wrong. Python did exactly what you told it to do, but what you told it to do wasn't what you meant.

* **Example:** You want to calculate the average of two numbers but forget about the order of operations.
* **The Code:**
    ```python
    # Goal: Average of 10 and 20 (Should be 15)
    result = 10 + 20 / 2 
    print(result) # Output: 20.0 (Because of PEMDAS, it did 20/2 first)
    ```
* **How to Debug:**
    1.  **Rubber Duck Debugging:** Explain your code line-by-line to an object (or a friend). Often, as you describe what the code *should* do, you’ll realize where the logic fails.
    2.  **Verify Assumptions:** Test small parts of your math or logic separately.
    3.  **Use a Debugger:** Use the built-in debugger in your code editor to "Step Through" the code line-by-line and watch how the variable values change in real-time.

---

#### **Comparison Summary**

| Error Type | When it occurs | Does it run? | Analogy |
| :--- | :--- | :--- | :--- |
| **Syntax** | Before running | No | Using a comma instead of a period in a sentence. |
| **Runtime** | While running | Yes (until crash) | Following a recipe but finding you're out of milk halfway through. |
| **Semantic** | While running | Yes | Asking for directions to the "bank" but being taken to a river bank. |

## **Glossary**

- **variable:** A name that refers to a value.
- **assignment statement:** A statement that assigns a value to a variable.
- **state diagram:** A graphical representation of a set of variables and the values they refer to.
- **keyword:** A special word used to specify the structure of a program.
- **import statement:** A statement that reads a module file and creates a module object.
- **module:** A file that contains Python code, including function definitions and sometimes other statements.
- **dot operator:** The operator, ., used to access a function in another module by specifying the module name followed by a dot and the function name.
- **statement:** One or more lines of code that represent a command or action.
- **evaluate:** Perform the operations in an expression in order to compute a value.
- **execute:** Run a statement and do what it says.
- **argument:** A value provided to a function when the function is called. Each argument is assigned to the corresponding parameter in the function.
- **comment:** Text included in a program that provides information about the program but has no effect on its execution.
- **runtime error:** An error that causes a program to display an error message and exit.
- **exception:** An error that is detected while the program is running.
- **semantic error:** An error that causes a program to do the wrong thing, but not to display an error message.

#### **Why is class a keyword?**

In Python, a **keyword** is a reserved word that has a predefined meaning and specific rules associated with it. You cannot use these words as variable names, function names, or any other identifier because the Python interpreter uses them to understand the **structure** of your code.

`class` is a keyword because it is the fundamental building block for **Object-Oriented Programming (OOP)**.

---

- **1. Why `class` Must Be a Keyword**
    - If `class` weren't a keyword, the interpreter wouldn't know when you are trying to define a new data type versus when you are just naming a variable.

- **The Interpreter's Perspective**
    - When Python sees the word `class`, it prepares to:
        1.  Create a new **namespace**.
        2.  Execute the code block within that namespace.
        3.  Bind that namespace to a new **type object**.



---

- **2. What Happens if You Try to Use it as a Variable?**
    - Because it is a keyword, Python will stop you immediately with a **Syntax Error** if you try to repurpose it.

```python
# This will crash your program
class = "Math 101" 
# Output: SyntaxError: invalid syntax
```

To get around this (for example, if you are naming a CSS class in a script), the Pythonic convention is to add a **trailing underscore**:
```python
css_class = "button-primary" # Preferred
class_ = "Math 101"          # Also acceptable
```

---

- **3. The Functional Role of `class`**
    - The `class` keyword acts as a **blueprint**. Without it, you would have to manually create dictionaries and functions to mimic objects, which would be messy and inefficient.

| Feature | Using the `class` Keyword | Without `class` (Manual) |
| :--- | :--- | :--- |
| **Organization** | Bundles data and logic together cleanly. | Data and functions are scattered. |
| **Inheritance** | Easy to reuse code from other classes. | Requires complex manual copying. |
| **Consistency** | Every "instance" follows the same rules. | Hard to ensure every "object" is the same. |

---

- **4. Summary of Keywords**
    - Keywords like `class`, `def`, `if`, and `while` are the "verbs" and "connectors" of the Python language. You can see the full list of keywords in your own terminal by running:

```python
import keyword
print(keyword.kwlist)
```

#### **Why can’t variable names be keywords?**

Think of Python as a **translator**. It takes the English-like words you write and converts them into machine code that a computer processor can understand.

For this translation to be fast and accurate, Python needs a set of **"Reserved Words"** (Keywords) that have a fixed, unchangeable meaning. If you were allowed to use these as variable names, the "translator" would get hopelessly confused.

---

- **1. The "Ambiguity" Problem**
    - Imagine if Python allowed you to name a variable `if`. Look at this hypothetical code:

```python
if = 10
if if == 10:
    print("Is this the variable or the command?")
```

The computer wouldn't know if you are trying to **assign a value** or **start a conditional logic block**. By making `if`, `class`, `def`, and `while` off-limits for variables, Python ensures that every line of code has one (and only one) clear interpretation.

---

- **2. Technical Reason: The Tokenizer**
    - When you run a Python script, the first thing that happens is **Lexical Analysis**. A part of the interpreter called the "Tokenizer" breaks your code into small chunks (tokens).

* **Keywords** are tagged as "Instruction Tokens."
* **Variable names** are tagged as "Identifier Tokens."

If a word could be both, the Tokenizer would essentially "short-circuit." It needs to know immediately upon seeing the word `for` that a loop is starting, without having to scan the rest of the file to see if you defined `for = 5` somewhere else.



---

- **3. Real-World Analogy: Traffic Signs**
    - Imagine you are driving and you see a **STOP** sign. In the world of driving, "STOP" is a keyword. 

If a local business were allowed to put up a billboard that was shaped like a red octagon and just said "STOP (at our cafe!)", drivers would be slamming on their brakes constantly. To keep traffic moving safely, the government **reserves** certain shapes and words for specific instructions. Python does the same for your code's "traffic."

---

- **4. How to Check for Keywords**
    - If you aren't sure if a name is "taken," you can check it instantly in Python:

```python
import keyword

print(keyword.iskeyword("class")) # True
print(keyword.iskeyword("my_var")) # False
```

- **The "Pro" Workaround**
    - If you absolutely *must* use a name that is a keyword (common when working with databases or APIs that use words like `type` or `id`), the standard convention is to add a **trailing underscore**:

* `class_`
* `type_`
* `global_`

This satisfies the interpreter while keeping your meaning clear to other humans.

#### **Why is it bad to use int, float, and string as variable names?**

Technically, Python **will** let you use `int`, `float`, and `str` as variable names. Unlike `class` or `if`, these are not **keywords**; they are **built-in functions** (or types). 

However, doing this is considered a "cardinal sin" in Python programming. Here is why it’s a bad idea and how it can break your code.

---

- **1. You "Shadow" the Built-in Function**
    - When you assign a value to `int`, you are telling Python: *"From now on, the name 'int' refers to this number, not the function that converts things into integers."* This is called **Shadowing**.

- The Problem in Action:
```python
int = 5  # You just "broke" the built-in int function
print(int) # This works fine and prints 5

# Later in your code...
age = int("25") # CRASH! 
```
**The Error:** `TypeError: 'int' object is not callable`
Python is trying to treat the number `5` like a function. It’s as if you renamed your hammer "Screwdriver"—the next time you actually need a screwdriver, you're going to be very frustrated.

---

- **2. It Confuses Other Programmers**
    - Python code is meant to be readable. When a developer sees `str(data)`, they immediately know you are converting a value to a string. 

If you have a variable named `str`, they have to look back through your code to figure out what `str` actually represents. It ruins the **predictability** of the language.

---

- **3. Breaking Standard Libraries**
    - Many Python libraries and modules rely on these built-in types to function. If you shadow a built-in name in a global scope, you might accidentally break a function inside a library that expects `float` to mean "a floating-point constructor."



---

- **4. How to Spot "Taken" Names**
    - Most Code Editors (VS Code, PyCharm, etc.) will change the color of built-in names.
        * **Blue/Purple:** Usually a keyword or built-in (Avoid using these!).
        * **White/Light Blue:** Usually a standard variable name (Safe to use!).

- **The "Pro" Solution**
    - If you really need a variable to represent an integer or a string, follow the same convention we use for keywords: **add a trailing underscore.**

| Bad Name | Better Name | Best Name (Contextual) |
| :--- | :--- | :--- |
| `int = 10` | `int_ = 10` | `user_age = 10` |
| `str = "Hi"` | `str_ = "Hi"` | `message = "Hi"` |
| `list = [1,2]` | `list_ = [1,2]` | `scores = [1,2]` |

---

- **Summary**
    - Using built-ins as variables creates **Semantic Errors**. Your code might look fine, but it will fail later in unexpected ways because you've removed a core tool from Python's toolbox.

Have you ever run into a `TypeError: 'list' object is not callable`? That’s almost always a sign that a variable was named `list` earlier in the script!

- **3. Comparison: Built-in vs. Math Module**
    - There is some overlap, but the `math` module functions are generally more specialized or always return floats.

| Task | Built-in | Math Module | Difference |
| :--- | :--- | :--- | :--- |
| **Power** | `pow(2, 3)` | `math.pow(2, 3)` | `math.pow` always returns a float (`8.0`). |
| **Absolute** | `abs(-5)` | `math.fabs(-5)` | `math.fabs` always returns a float (`5.0`). |
| **Rounding** | `round(2.5)` | `math.floor(2.5)` | `round` follows "Banker's rounding"; `floor` always goes down. |

#### **what modules are considered core Python?**

When developers refer to "Core Python," they usually mean the **Python Standard Library**. This is the collection of modules that comes "pre-installed" with Python. You don't need to use `pip install`; you just `import` them.

Because the standard library is massive (hundreds of modules), it is helpful to categorize the most essential "core" modules by their purpose.

---

- **1. System and File Interaction**
    - These modules allow Python to talk to your computer’s operating system.

* **`os`**: Used for operating system dependent functionality like reading/writing files, joining paths, and listed directories.
* **`sys`**: Provides access to variables and functions that interact strongly with the Python interpreter (e.g., `sys.argv` for command-line arguments).
* **`shutil`**: High-level file operations like copying, moving, and deleting entire folders.
* **`pathlib`**: The modern way to handle file paths as objects rather than strings.

---

- **2. Data Processing and Handling**
    - These are the "workhorse" modules for managing information.

* **`json`**: The go-to module for encoding and decoding JSON data (essential for web APIs).
* **`csv`**: For reading and writing tabular data in CSV format.
* **`re`**: Provides regular expression matching operations for advanced string searching and manipulation.
* **`collections`**: Offers specialized container datatypes (like `namedtuple`, `deque`, and `Counter`) that go beyond standard lists and dicts.

---

- **3. Math and Science**
    - While built-in functions handle basics, these provide the heavy lifting.

* **`math`**: Standard mathematical functions (trigonometry, logarithms, constants like $\pi$).
* **`random`**: Used for generating pseudo-random numbers, shuffling sequences, or picking random elements.
* **`statistics`**: Basic statistical functions like `mean`, `median`, and `stdev`.
* **`decimal`**: Essential for financial applications where floating-point math (like `0.1 + 0.2`) needs to be perfectly precise.

---

- **4. Date and Time**
    - Handling time is notoriously difficult in programming; these modules make it manageable.

* **`datetime`**: The primary module for manipulating dates and times.
* **`time`**: Used for time-related tasks, like pausing execution (`time.sleep()`) or getting the current Unix timestamp.

---

- **5. Web and Networking**
    - Even without external frameworks like Django or Flask, Python has core networking capabilities.

* **`urllib`**: A package for handling URLs and making network requests.
* **`http`**: Low-level HTTP client and server implementation.
* **`socket`**: The foundation of all network communication in Python.

---

- **6. Debugging and Development**
    - Tools that help you write better code and find bugs.

* **`logging`**: A flexible system for tracking events that happen when your software runs (much better than using `print()`).
* **`unittest`**: The built-in framework for writing and running automated tests.
* **`pydoc`**: Automatically generates documentation from your Python modules.



---

- **The "Batteries Included" Philosophy**
    - Python is often described as a **"Batteries Included"** language. This means that for about 80% of common programming tasks—from parsing a website to calculating standard deviation—the "core" modules have everything you need without ever needing to download anything from the internet.
